```python
# 1. Bank Account System with Custom Exceptions


class InsufficientBalance(Exception):
    pass


class BankAccount:

    def __init__(self, name, balance):
        self.name = name
        self.balance = balance

    def deposit(self, amount):
        self.balance += amount
        print("Deposited:", amount)

    def withdraw(self, amount):
        if amount > self.balance:
            raise InsufficientBalance("Insufficient balance")
        self.balance -= amount
        print("Withdrawn:", amount)

    def transfer(self, other, amount):
        self.withdraw(amount)
        other.deposit(amount)
        print("Transfer successful")

    def show_balance(self):
        print("Balance:", self.balance)


a1 = BankAccount("Rahul", 50000)
a2 = BankAccount("Amit", 20000)

a1.deposit(5000)
a1.withdraw(10000)
a1.transfer(a2, 15000)

a1.show_balance()
a2.show_balance()
```

```python
# 2. Employee Salary Management



class Employee:

    def __init__(self, name, salary):
        self.name = name
        self.salary = salary

    def calculate_salary(self):
        return self.salary


class Manager(Employee):

    def calculate_salary(self):
        bonus = self.salary * 0.20
        return self.salary + bonus


class Developer(Employee):

    def calculate_salary(self):
        bonus = self.salary * 0.10
        return self.salary + bonus


e1 = Employee("Raj", 40000)
e2 = Manager("Amit", 60000)
e3 = Developer("Ravi", 50000)

print(e1.name, e1.calculate_salary())
print(e2.name, e2.calculate_salary())
print(e3.name, e3.calculate_salary())
```

```python
# 3. Shopping Cart System


class Product:

    def __init__(self, name, price, quantity):
        self.name = name
        self.price = price
        self.quantity = quantity

    def total_price(self):
        return self.price * self.quantity


class ShoppingCart:

    def __init__(self):
        self.products = []

    def add_product(self, product):
        self.products.append(product)

    def remove_product(self, name):
        for product in self.products:
            if product.name == name:
                self.products.remove(product)

    def total(self):
        total = 0

        for product in self.products:
            total += product.total_price()

        return total

    def final_bill(self, discount):
        total = self.total()
        discount_amount = total * discount / 100
        return total - discount_amount


p1 = Product("Laptop", 50000, 1)
p2 = Product("Mouse", 1000, 2)

cart = ShoppingCart()

cart.add_product(p1)
cart.add_product(p2)

print("Total:", cart.total())
print("Final Bill:", cart.final_bill(10))
```

```python
# 4. Vehicle Rental System


class Vehicle:

    def __init__(self, name):
        self.name = name

    def rental_cost(self, days):
        return 0


class Car(Vehicle):

    def rental_cost(self, days):
        return days * 2000


class Bike(Vehicle):

    def rental_cost(self, days):
        return days * 800


car = Car("Toyota")
bike = Bike("Royal Enfield")

print(car.name)
print("Car Rent:", car.rental_cost(5))

print(bike.name)
print("Bike Rent:", bike.rental_cost(5))
```

```python
# 5. Online Examination System


class Question:

    def __init__(self, question, options, answer):
        self.question = question
        self.options = options
        self.answer = answer

    def check_answer(self, user_answer):
        return user_answer == self.answer


class Student:

    def __init__(self, name):
        self.name = name
        self.score = 0

    def attempt_exam(self, questions, answers):

        for i in range(len(questions)):

            if questions[i].check_answer(answers[i]):
                self.score += 1

    def show_result(self, total_questions):

        print("Student:", self.name)
        print("Score:", self.score, "/", total_questions)


q1 = Question(
    "Which language is used for Data Science?",
    ["Java", "Python", "HTML", "CSS"],
    "Python"
)

q2 = Question(
    "What is 10 + 20?",
    ["20", "30", "40", "50"],
    "30"
)

questions = [q1, q2]

student = Student("Rahul")

answers = ["Python", "30"]

student.attempt_exam(questions, answers)

student.show_result(len(questions))
```
