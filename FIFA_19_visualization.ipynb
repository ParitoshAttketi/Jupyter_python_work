```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

```

```python
df= pd.read_csv("C:\\Users\\bhoom\\Downloads\\FIFA_19_cleared_data.csv")
df
```

```python
df.columns
```

```python
df["Work_Rate_Attack"].value_counts().plot(kind="bar")
```

```python
df["Work_Rate_Defense"].value_counts().plot(kind="bar")
```

```python
# avg Age players

plt.figure(figsize=(8, 5))

sns.histplot(
    data=df,
    x="Age",
    bins= 10
)
```

```python
# players by prefered foot

df["Preferred Foot"].value_counts().plot(kind="bar")
```

```python
# number of players positon 

df["Position"].value_counts().plot(kind="bar")
```

```python
# Top 10 Nationality by number of players

df["Nationality"].value_counts().head(10).plot(kind="bar")
```

```python
# top 10 clubs by the number of players

df["Club"].value_counts().head(10).plot(kind="bar")
```

```python
# Overall 

df["Overall"].plot(kind="hist", bins=20)

plt.title("Players Overall Rating Avg")
plt.xlabel("Overall rating")
plt.ylabel("player count")
plt.show()
```

```python
# Potential

df["Potential"].plot(kind="hist", bins=20)
plt.title("Players Potential Overall Rating Avg")
plt.xlabel("Overall rating")
plt.ylabel("player count")
plt.show()
```

```python
# Overall vs Potential

df.plot(x="Overall", y="Potential", kind="scatter")
plt.title("Players Potential vs Overall Rating Avg")
plt.xlabel("Overall rating")
plt.ylabel("player count")
plt.show()
```

```python
# players overall vs value in euro

df.plot(x="Overall", y="Value in Euro", kind="scatter")
plt.show()
```

```python
# players overall vs wage

df.plot(x="Overall", y="Wage in Euro", kind="scatter")
plt.show()
```

```python
# Avg overall by prefered foot

df.groupby("Preferred Foot")["Overall"].mean().plot(kind="bar")
plt.show()
```

```python
# Avg overall by position 

df.groupby("Position")["Overall"].mean().sort_values().plot(kind="bar")
plt.show()
```

```python
# Avg Value by work rate attack

df.groupby("Work_Rate_Attack")["Value in Euro"].mean().plot(kind="bar")
plt.show()
```
