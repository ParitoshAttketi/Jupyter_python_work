```python
import pandas as pd

df = pd.DataFrame({
    "Name": ["Amit", "Rahul", "Priya", "Sneha", "Vikram", "Anjali", "Kiran", "Rohit"],
    "Age": [25, 30, 28, 24, 35, 29, 31, 27],
    "City": ["Hyderabad", "Mumbai", "Delhi", "Hyderabad", "Mumbai", "Delhi", "Hyderabad", "Mumbai"],
    "Department": ["IT", "HR", "IT", "Finance", "IT", "HR", "Finance", "IT"],
    "Salary": [45000, 55000, 60000, 40000, 75000, 50000, 48000, 65000],
    "Experience": [2, 5, 4, 1, 8, 3, 6, 5]
})

df
```

```python
# 12. Find employees whose salary is greater than 50000 AND experience is greater than 4.

df[(df["Salary"] > 50000) & (df["Experience"] > 4)]
```

```python
# 13. Find the average salary.

df["Salary"].mean()
```

```python
# 14. Find the highest salary.

df["Salary"].max()
```

```python
# 15. Find the employee who has the highest salary.

df[df["Salary"] == df["Salary"].max()]
```
