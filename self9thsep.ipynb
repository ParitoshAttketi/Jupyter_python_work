# 1. Remove duplicates from a list without set()

```python
numbers = [10, 20, 10, 30, 20, 40]

result = []

for x in numbers:
    if x not in result:
        result.append(x)

print(result)
```

# 2. Find the second-largest number

```python
numbers = [10, 50, 30, 50, 20, 40]

unique = list(set(numbers))
unique.sort()

print(unique[-2])
```

# 3. Count word frequency

```python
text = "python is easy python is powerful"

words = text.split()
count = {}

for word in words:
    count[word] = count.get(word, 0) + 1

print(count)
```

# 4. Find common elements

```python
a = [10, 20, 30, 40]
b = [30, 40, 50, 60]

common = []

for x in a:
    if x in b:
        common.append(x)

print(common)
```

# 5. Find the highest salary

```python
employees = {
    "Rahul": 45000,
    "Priya": 60000,
    "Amit": 52000
}

highest = max(employees, key=employees.get)

print(highest)
print(employees[highest])
```

# 6. Use a lambda function for sorting

```python
students = [
    ("A", 85),
    ("B", 72),
    ("C", 95),
    ("D", 60)
]

students.sort(key=lambda x: x[1])

print(students)
```

# 7. Find even numbers using list comprehension

```python
numbers = [1, 2, 3, 4, 5, 6]

result = [x**2 for x in numbers if x % 2 == 0]

print(result)
```

# 8. Create a function for prime numbers

```python
def prime(n):
    if n < 2:
        return False

    for i in range(2, n):
        if n % i == 0:
            return False

    return True

print(prime(7))
```

# 9. Find duplicate values

```python
numbers = [10, 20, 10, 30, 20, 40]

duplicates = []

for x in numbers:
    if numbers.count(x) > 1 and x not in duplicates:
        duplicates.append(x)

print(duplicates)
```

# 10. Reverse a dictionary

```python
data = {
    "A": 1,
    "B": 2,
    "C": 3
}

result = {}

for key, value in data.items():
    result[value] = key

print(result)
```

# 11. Find the most frequent number

```python
numbers = [1, 2, 2, 3, 2, 4, 3]

most = max(numbers, key=numbers.count)

print(most)
```

# 12. Handle exceptions

```python
a = 10
b = 0

try:
    result = a / b
    print(result)

except ZeroDivisionError:
    print("Cannot divide by zero")
```

# 13. Create a class

```python
class Student:

    def __init__(self, name, marks):
        self.name = name
        self.marks = marks

    def display(self):
        print(self.name)
        print(self.marks)


s1 = Student("Rahul", 85)

s1.display()
```

# 14. Use inheritance

```python
class Vehicle:

    def start(self):
        print("Vehicle started")


class Car(Vehicle):

    def drive(self):
        print("Car is driving")


c = Car()

c.start()
c.drive()
```

# 15. Create a generator

```python
def numbers():
    for i in range(1, 6):
        yield i


for x in numbers():
    print(x)
```

# 16. Find missing values in pandas

```python
import pandas as pd

df = pd.DataFrame({
    "Name": ["A", "B", "C"],
    "Age": [20, None, 25],
    "Salary": [30000, 40000, None]
})

print(df.isna().sum())
```

# 17. Fill missing values

```python
df["Age"] = df["Age"].fillna(df["Age"].mean())

print(df)
```

# 18. Group data using pandas

```python

df = pd.DataFrame({
    "Category": ["A", "B", "A", "B"],
    "Sales": [100, 200, 150, 300]
})

result = df.groupby("Category")["Sales"].sum()

print(result)
```

# 19. Extract year from a date

```python
df = pd.DataFrame({
    "Order_Date": ["2025-01-10", "2025-05-20", "2026-02-15"]
})

df["Order_Date"] = pd.to_datetime(df["Order_Date"])

df["Year"] = df["Order_Date"].dt.year

print(df)
```

# 20. Find the highest-selling product

```python
df = pd.DataFrame({
    "Product": ["A", "B", "A", "C", "B"],
    "Sales": [100, 200, 150, 300, 100]
})

result = df.groupby("Product")["Sales"].sum()

print(result.idxmax())
print(result.max())
```
