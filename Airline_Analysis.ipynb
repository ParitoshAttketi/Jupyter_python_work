## Duration

```python
df["Duration"].isna().sum()
```

```python
df["Duration"].unique()
```

```python
df["Duration"] = (
    df["Duration"]
    .str.replace("h ", ":", regex=False)
    .str.replace("h", ":00", regex=False)
    .str.replace("m", "", regex=False)
)
```

```python
df["Duration"].unique()
```

```python
df["Duration"]= pd.to_datetime(df["Duration"],errors ='coerce')
```

```python
df["Duration"] = df["Duration"].dt.strftime("%H:%M")
```

```python
df["Duration_HR"] = df["Duration"].str.split(":").str[0]
df["Duration_MIN"] = df["Duration"].str.split(":").str[1]
```

```python
df["Duration_HR"] = pd.to_numeric(df["Duration_HR"],errors = 'coerce')
```

```python
df["Duration_HR"].fillna(0,inplace = True)
```

```python
df["Duration_HR"].astype('int')
```

```python
df["Duration_MIN"].fillna(0,inplace = True)
```

```python
df["Duration_MIN"].astype('int')
```
