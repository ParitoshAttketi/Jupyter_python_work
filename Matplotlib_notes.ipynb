```python
import matplotlib.pyplot as plt
import numpy as np
```

```python
x= [1,2,3,4,5]
y= [10,11,12,13,14]

plt.plot(x,y)
plt.show
```

```python
# with label

plt.plot(x,y,
         color = "blue",
         linewidth = 1,
         marker = ".",
         markersize = 8,
         markeredgecolor = "Red",
         markerfacecolor = "Black"
        )

plt.title("First plot")
plt.xlabel(" X axis ")
plt.ylabel(" Y axis ")
plt.show
```

```python
# pyplot common command

X = np.linspace(0,10,100)
Y = np.sin(X)
```

```python
plt.plot(X,Y,
         color = "blue",
         linewidth = 1,
         marker = ".",
         markersize = 8,
         markeredgecolor = "Red",
         markerfacecolor = "Black"
        )
plt.title("Sine waves ")
plt.xlabel(" X axis data ")
plt.ylabel(" sin X axis ")
plt.xlim(1,10)
plt.ylim(-1.5,1.5)
plt.show
```

```python
plt.plot(X,Y,
         color = "blue",
         linewidth = 1,
         marker = ".",
         markersize = 8,
         markeredgecolor = "Red",
         markerfacecolor = "Black"
        )
plt.title("Sine waves ")
plt.xlabel(" X axis data ")
plt.ylabel(" sin X axis ")
plt.xlim(1,10)
plt.ylim(-1.5,1.5)
plt.xticks([0,2,4,6,8,10])
plt.yticks([-1,0,1])
plt.grid(True, alpha=0.5)
plt.legend(["Sine waves"])
plt.tight_layout()
plt.show
```

```python
# multiplot

X= np.linspace(0,10,100)


# create the multiplot figure
fig,axes = plt.subplots(nrows = 2,
                       ncols = 2,
                        figsize=(10,6))


# fig 
fig.set_size_inches(8,8) # resize fig in inches
fig.set_facecolor("beige") # change the background color
fig.suptitle("Figure Title", 
             fontsize=16,
            fontweight="bold"
            )

axes[0,0].plot(X, np.sin(X), 'blue')
axes[0,0].set_title("Shine")

axes[0,1].plot(X, np.cos(X), 'blue')
axes[0,1].set_title("cos plot")

axes[1,0].plot(X, np.tanh(X), 'gray')
axes[1,0].set_title("tanh plot")

axes[1,1].plot(X, X**2, 'tan')
axes[1,1].set_title("Squred plot")
plt.show
```

### Legend method:

```python
# create a plot:

x= np.linspace(0,15,200)
y= np.sin(x)
# create the figure
plt.figure(figsize = (10,8))

# plot the data
plt.plot(x,y,color = "skyblue",
         marker = "8",
         markeredgecolor ="black",
         markerfacecolor ="green",
         markersize = 3
        )

# add the legend
plt.legend(["sin"],
           loc = "lower right",
           fontsize = 15,
           title = "Info",
           title_fontsize = 15,
           edgecolor ="black",
           facecolor = "lightgreen",
           shadow = True
          )

plt.show
```

```python
# 

months = ["jan","feb","mar","apr","may","jun","jul","aug","sep","oct","nov","dec"]
product_a =[120,135,148,162,155,178,195,210,198,220,235,250]
product_b =[80,95,88,105,112,120,135,128,142,155,160,175]
product_c =[60,65,70,75,85,90,95,100,108,115,120,130]

# make the plot
X = range(len(months))

# make the sub plots
fig, ax=plt.subplots(2,2,
                     figsize=(12,6))

# make the plot
ax[0,0].plot(X,product_a,
       color = 'purple',
       linewidth = 2,
       marker = "o",
       markersize = 6,
       label = "Product A")
ax[0,0].set_title("Product A")
ax[0,0].legend(["Product A"])

ax[0,1].plot(X,product_b,
       color = 'lightgreen',
       linewidth = 2,
       marker = "o",
       markersize = 6,
       label = "Product B")
ax[0,1].set_title("Product B")
ax[0,1].legend(["Product B"])

ax[1,0].plot(X,product_c,
       color = 'lightblue',
       linewidth = 2,
       marker = "o",
       markersize = 6,
       label = "Product C")
ax[1,0].set_title("Product C")
ax[1,0].legend(["Product C"])

ax[1,1].plot(product_a,product_b,
       color = 'brown',
       linewidth = 2,
       marker = "o",
       markersize = 6,
       label = "Product A"
       )
ax[1,1].set_title("Product B vs A")
ax[1,1].legend(["Product B vs A"])

plt.show()
```

```python
# Department salary data 
departments = ['Engineering', 'Marketing', 
               'Sales', 'HR', 'Finance'] 
avg_salary  = [95000, 75000, 70000, 
               65000, 85000] 
max_salary  = [150000, 110000, 120000, 
               90000, 130000] 

# make the subplots
fig,ax = plt.subplots(1,2,figsize = (12,8))

x = np.arange(len(departments)) 
width = 0.35     # bar width 
# make the plot 

# Two sets of bars 
ax[0].bar(x - width/2, 
               avg_salary, 
               width, 
               label='Average Salary', 
               color='purple', 
               edgecolor='black', 
               linewidth=0.5)
ax[0].set_ylim(10000,90000)

ax[1].bar(x + width/2, 
               avg_salary, 
               width, 
               label='Average Salary', 
               color='lightgreen', 
               edgecolor='black', 
               linewidth=0.5)
```
