```python
# generator
# syntax

def generator_name():
    yield val1
    yield val2
    yield val3

g = generator_name()
print(next(g))
```

```python

def lists():
    yield "paritosh"
    yield 2
    yield 3
    yield 4

g = lists()

print(next(g))
print(next(g))
print(next(g))
print(next(g))
```

```python
def loop():
    for i in range(1,6):
        yield i

g = loop()

for i in g:
    print(i)
```
