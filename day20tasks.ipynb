```python
# 1. Create two parent classes Father and Mother. Create a child class Child that inherits from both.

class Father:
    def __init__(self, name, **kwargs):
        super().__init__(**kwargs)
        self.father_name = name 

class Mother:
    def __init__(self, name2, **kwargs):
        super().__init__(**kwargs)
        self.mother_name = name2

class Child(Father, Mother):
    def __init__(self, name, name2, name3):
        super().__init__(name=name, name2=name2)
        self.child_name = name3

    def display(self):
        print(f"Father name : {self.father_name}")
        print(f"Mother name : {self.mother_name}")
        print(f"Child name : {self.child_name}")

c = Child("Santosh", "Paritosh", "Sai")
c.display()
```

```python
# 13. Create class Validator with static method is_valid(). Create class Form that inherits and uses it.

class Validator:
    @staticmethod
    def is_valid(name):
        return len(name) >= 3


class Form(Validator):
    def submit(self, name):
        if self.is_valid(name):
            print("Form submitted successfully.")
        else:
            print("Invalid input. Name must be at least 3 characters.")


obj = Form()

obj.submit("Paritosh")
obj.submit("Al")
```

```python
# 14. Define class Converter with static method to_celsius(). Create class WeatherApp that
# inherits from multiple classes including Converter.

class Converter:
    @staticmethod
    def to_celsius(fahrenheit):
        return (fahrenheit - 32) * 5 / 9


class Location:
    def show_location(self):
        print("Location: Hyderabad")


class WeatherInfo:
    def show_weather(self):
        print("Today's Weather: Sunny")


class WeatherApp(Location, WeatherInfo, Converter):
    def display_temperature(self, fahrenheit):
        celsius = self.to_celsius(fahrenheit)
        print(f"Temperature: {fahrenheit}°F = {celsius:.2f}°C")


obj = WeatherApp()

obj.show_location()
obj.show_weather()
obj.display_temperature(95)
```

```python
# 19. Create two abstract classes Shape and Color. Create a concrete class that inherits from
# both and implements abstract methods.

from abc import ABC, abstractmethod

class Shape(ABC):
    @abstractmethod
    def area(self):
        pass


class Color(ABC):
    @abstractmethod
    def show_color(self):
        pass


class Rectangle(Shape, Color):
    def __init__(self, length, width, color):
        self.length = length
        self.width = width
        self.color = color

    def area(self):
        return self.length * self.width

    def show_color(self):
        print("Color:", self.color)


obj = Rectangle(10, 5, "Blue")

print("Area =", obj.area())
obj.show_color()
```

```python
# 26. (Problem) Create two parent classes:
# • Employee with method get_salary()
# • Person with method get_details() Create class Manager that inherits from both. Add
# a method that combines both parent methods.

class Employee:
    def __init__(self, salary):
        self.salary = salary

    def get_salary(self):
        return self.salary


class Person:
    def __init__(self, name, age):
        self.name = name
        self.age = age

    def get_details(self):
        return f"Name: {self.name}, Age: {self.age}"


class Manager(Employee, Person):
    def __init__(self, name, age, salary):
        Employee.__init__(self, salary)
        Person.__init__(self, name, age)

    def manager_info(self):
        print(self.get_details())
        print("Salary:", self.get_salary())


obj = Manager("Paritosh", 25, 80000)
obj.manager_info()
```

```python
# 28. (Problem) Design a system using multiple inheritance:
# • Class Writer
# • Class Editor Create class Author that inherits from both. Add static method to count total authors.

class Writer:
    def write(self):
        print(f"{self.name} is writing a book.")


class Editor:
    def edit(self):
        print(f"{self.name} is editing the script.")


class Author(Writer, Editor):
    total_authors = 0

    def __init__(self, name):
        self.name = name
        Author.total_authors += 1

    @staticmethod
    def count_authors():
        print("Total Authors:", Author.total_authors)

a1 = Author("Paritosh")
a2 = Author("Rahul")

a1.write()
a1.edit()

a2.write()
a2.edit()
Author.count_authors()
```

```python
# 29. (Problem) Create two abstract classes Printable and Scannable. Create class
# PrinterScanner that inherits from both and implements both abstract methods.

class Printable(ABC):
    @abstractmethod
    def print_document(self):
        pass


class Scannable(ABC):
    @abstractmethod
    def scan_document(self):
        pass


class PrinterScanner(Printable, Scannable):
    def print_document(self):
        print("Printing the document")

    def scan_document(self):
        print("Scanning the document")


obj = PrinterScanner()

obj.print_document()
obj.scan_document()
```
