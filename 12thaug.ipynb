```python
# 1. Find the second largest number

numbers = [10, 25, 8, 45, 30, 45]

numbers = list(set(numbers))
numbers.sort()

print(numbers[-2])
```

```python
# 2. Count frequency of elements

numbers = [1, 2, 2, 3, 3, 3, 4]

count = {}

for n in numbers:
    count[n] = count.get(n, 0) + 1

print(count)
```

```python
# 3. Remove duplicate values


numbers = [10, 20, 10, 30, 20, 40]

result = list(set(numbers))

print(result)
```

```python
# 4. Find common elements

a = [1, 2, 3, 4, 5]
b = [3, 4, 5, 6, 7]

common = list(set(a) & set(b))

print(common)
```

```python
# 5. Reverse words in a sentence


sentence = "Python is very easy"

words = sentence.split()
words.reverse()

print(" ".join(words))
```

```python
# 6. Find vowels and consonants


text = "Python Programming"

vowels = 0
consonants = 0

for ch in text.lower():
    if ch in "aeiou":
        vowels += 1
    elif ch.isalpha():
        consonants += 1

print("Vowels:", vowels)
print("Consonants:", consonants)
```

```python
# 7. Find the largest value in a dictionary


marks = {
    "Rahul": 78,
    "Priya": 92,
    "Amit": 85,
    "Sneha": 88
}

student = max(marks, key=marks.get)

print(student)
print(marks[student])
```

```python
# 8. Sort a dictionary by values


marks = {
    "Rahul": 78,
    "Priya": 92,
    "Amit": 85
}

result = dict(sorted(marks.items(), key=lambda x: x[1]))

print(result)
```

```python
# 9. Find prime numbers


for n in range(2, 51):
    prime = True

    for i in range(2, n):
        if n % i == 0:
            prime = False
            break

    if prime:
        print(n)
```

```python
# 10. Create a function for factorial

def factorial(n):
    result = 1

    for i in range(1, n + 1):
        result = result * i

    return result


print(factorial(5))
```

```python
# 11. Find palindrome

text = "madam"

if text == text[::-1]:
    print("Palindrome")
else:
    print("Not Palindrome")
```

```python
# 12. Find missing number

numbers = [1, 2, 3, 4, 5, 6, 8, 9, 10]

total = sum(range(1, 11))
missing = total - sum(numbers)

print(missing)
```

```python
# 13. Use list comprehension


numbers = range(1, 21)

result = [x**2 for x in numbers if x % 2 == 0]

print(result)
```

```python
# 14. Find words longer than 5 characters

words = ["Python", "Java", "Programming", "SQL", "Machine", "AI"]

result = [word for word in words if len(word) > 5]

print(result)
```

```python
# 15. Handle division by zero


a = 10
b = 0

try:
    result = a / b
    print(result)

except ZeroDivisionError:
    print("Cannot divide by zero")
```

```python
# Task: Create a Student class with name and marks.

class Student:

    def __init__(self, name, marks):
        self.name = name
        self.marks = marks

    def display(self):
        print("Name:", self.name)
        print("Marks:", self.marks)


s1 = Student("Rahul", 85)

s1.display()
```

```python
# 17. Create a Bank Account class

class Bank:

    def __init__(self, balance):
        self.balance = balance

    def deposit(self, amount):
        self.balance += amount

    def withdraw(self, amount):
        if amount <= self.balance:
            self.balance -= amount
        else:
            print("Insufficient balance")

    def show_balance(self):
        print("Balance:", self.balance)


account = Bank(10000)

account.deposit(5000)
account.withdraw(3000)

account.show_balance()
```

```python
# 20. Filter employees by salary


employees = {
    "Rahul": 45000,
    "Priya": 65000,
    "Amit": 55000,
    "Sneha": 40000
}

result = {}

for name, salary in employees.items():
    if salary > 50000:
        result[name] = salary

print(result)
```
