```python
#  1. Create a hybrid inheritance structure with classes A → B and C → D (where D inherits
# from both B and C).
# Concept
# Hybrid Inheritance (Combination of Single + Multiple Inheritance)
# What to do
# • Create parent class A.
# • Create class B inheriting from A.
# • Create another class C.
# • Create class D inheriting from both B and C.

class A:
    def method(self):
        return "hello from A"

class B(A):
    def method_b(self):
        return "hello from B"

class C:
    def method_c(self):
        return "hello from C"

class D(B,C):
    def method_d(self):
        return "hello from D"

x = D()
print(x.method())
print(x.method_b())
print(x.method_c())
print(x.method_d())
```

```python
# 2. Create Person, Employee, Student, and TeachingAssistant using hybrid inheritance.
# Concept
# Real-world Hybrid Inheritance
# What to do
# • Person is the base class.
# • Employee inherits Person.
# • Student inherits Person.
# • TeachingAssistant inherits Employee and Student.

class Person:
    def method_a(self):
        return "hello from Person"

class Employee(Person):
    def method_b(self):
        return "hello from Employee"

class Student(Person):
    def method_c(self):
        return "hello from student"

class TeachingAssistant(Employee,Student):
    def method_d(self):
        return "hello from assistant"


x = TeachingAssistant()
print(x.method_a())
print(x.method_b())
print(x.method_c())
print(x.method_d())
```

```python
#3. Create Vehicle → Car, Electric → ElectricCar (ElectricCar inherits from both).
# Concept
# Hybrid Inheritance
# What to do
# • Vehicle is parent.
# • Car inherits Vehicle.
# • Electric is another parent.
# • ElectricCar inherits Car and Electric.

class Vehicle:
    def call(self):
        return "It is a vehicle"

class Car(Vehicle):
    def call_a(self):
        return "Car is a vehicle"

class Electric:
    def method(self):
        return "It is Electric"

class ElectricCar(Electric):
    def method_a(self):
        return "it is a car"
        
print("="*40)
x = Car()
print(x.call())
print(x.call_a())
print("="*40)
y = ElectricCar()
print(y.method())
print(y.method_a())
print("="*40)
```

```python
# 4 Demonstrate Hybrid Inheritance and print Method Resolution Order (MRO). Concept Method Resolution Order (MRO)
#1. Create four classes. 2. Give same method name in multiple classes. 3. Print ClassName.mro(). 4. Observe Python's searching order.

# Parent Class
class A:
    def show(self):
        print("show() from Class A")

class B(A):
    def show(self):
        print("show() from Class B")

class C:
    def show(self):
        print("show() from Class C")

class D(B, C):
    pass

obj = D()

obj.show()

print("Method Resolution Order (MRO):")
print(D.mro())
```

```python
# 5. Create class A, B, C, D and solve the Diamond Problem using super().
# 1. Create constructor in A.
#2. Call super() in B.
#3. Call super() in C.
#4. Call super() in D.
#5. Create object of D.
#6. Observe A constructor runs only once.


class A:
    def __init__(self):
        print("hello from A")

class B(A):
    def __init__(self):
        print("hello from B")
        super().__init__()

class C(A):
    def __init__(self):
        print("hello from C")
        super().__init__()

class D(B,C):
    def __init__(self):
        print("hello from D")
        super().__init__()

c = D()
```

```python
# 6. Create a banking system using hybrid inheritance.
# Hybrid + Real-world Example
# What to do
# Account
# |
# Savings
# Loan
# SavingsLoanAccount
# How to solve
# 1. Account stores account number.
# 2. Savings stores balance.
# 3. Loan stores loan amount.
# 4. SavingsLoanAccount inherits both.
# 5. Print all details.

class Account:
    def __init__(self,acc_no):
        self.acc_no = acc_no
        
class Savings(Account):
    def __init__(self,acc_no,balance):
        super().__init__(acc_no)
        self.balance = balance

class loan:
    def __init__(self,loan_amount):
        self.loan_amount = loan_amount

class SavingsLoanAccount(Savings,loan):
    def __init__(self,acc_no,balance,loan_amount):
        Savings.__init__(self,acc_no,balance)
        loan.__init__(self,loan_amount)
        
    def display(self):
        print("Savings Loan Account Details")
        print(f"Account no. : {self.acc_no}")
        print(f"Balance     : {self.balance}")
        print(f"loan        : {self.loan_amount}")

a=SavingsLoanAccount(12345,10000000,700000)
a.display()
        
```

```python
# 7. Create Shape, Color, Circle, and ColoredCircle.
# Concept
# Hybrid Inheritance
# What to do
# Shape
# |
# Circle
# Color
# ColoredCircle
# How to solve
# 1. Shape has area().
# 2. Color has color().
# 3. Circle inherits Shape.
# 4. ColoredCircle inherits Circle and Color.
# 5. Display area and color.

class Shape:
    def area(self):
        pass
        
class Circle(Shape):
    def __init__(self,radius):
        self.radius = radius
        
    def area(self):
        return 3.14 * (self.radius**2)
        
class Color:
    def __init__(self):
        self.colour = "Blue"

class ColoredCircle(Circle,Color):
    def __init__(self, radius):
        Circle.__init__(self, radius)
        Color.__init__(self)
        
    def display(self):
        print("Colored Circle Details")
        print(f"Area of the Circle   : {self.area()}")
        print(f"Colour of the circle : {self.colour}")

x = ColoredCircle(4)
x.display()
```

```python
# 8. Create constructors at every level.
# Concept
# Constructor Chaining
# What to do
# Every class has its own constructor.
# How to solve
# 1. Create constructor in every class.
# 2. Use super() everywhere.
# 3. Print constructor execution order.
# 4. Observe execution using MRO.

class A:
    def __init__(self):
        print("hello from A")

class B(A):
    def __init__(self):
        super().__init__()
        print("hello from b")

class C(A):
    def __init__(self):
        super().__init__()
        print("hello from c")

class D(B,C):
    def __init__(self):
        super().__init__()
        print("hello from d")

c = D()
    
print(D.mro())        
```

```python
# 9. Override methods in intermediate classes.
# Concept
# Method Overriding
# What to do
# A
# |
# B
# C
# D(B,C)
# All classes have
# show()
# How to solve
# 1. Create same method.
# 2. Override in B.
# 3. Override in C.
# 4. Call from D.
# 5. Observe MRO.

class A:
    def show(self):
        print("hello  from a")

class B(A):
    def show(self):
        print("hello from b")

class C:
    def show(self):
        print("hello from c")

class D(B,C):
    pass

x = D()

x.show()
print(D.mro())
```

```python
# 10 Create a Device → Phone, Camera → SmartPhone hierarchy.
# Concept
# Hybrid Inheritance
# What to do
# Device
# |
# Phone
# Camera
# SmartPhone
# How to solve
# 1. Device has power_on().
# 2. Phone has call().
# 3. Camera has click().
# 4. SmartPhone accesses all methods.

class Device:
    def power_on(self):
        print("turning on")

class Phone(Device):
    def call(self):
        print("phone is turning on")

class Camara:
    def click(self):
        print("Clicks when we use camara")

class SmartPhone(Phone,Camara):
    def display(self):
        print("All details")
        
x = SmartPhone()
x.display()
x.power_on()
x.call()
x.click()

print(SmartPhone.mro())  
```

```python
# 11. What is Duck Typing? Explain with a simple example.
# Concept
#Duck Typing
#"If an object behaves like a duck, Python treats it like a duck."
#What to do
#Create two unrelated classes having the same method.
#How to solve
#1. Create Dog.
#2. Create Robot.
#3. Both have speak().
#4. Write one function calling obj.speak().
#5. Pass both objects.

class Dog:
    def speak(self):
        print("Dog says: Woof!")

class Robot:
    def speak(self):
        print("Robot says: Beep Beep!")

def make_sound(obj):
    obj.speak()

dog = Dog()
robot = Robot()

make_sound(dog)
make_sound(robot)
```

```python
#12. Create Dog and Robot both having speak().
#Concept
#Duck Typing
#What to do
#Common method
#speak()
#How to solve
#1. Dog → speak().
#2. Robot → speak().
#3. Function accepts any object.
#4. Call obj.speak().

class Dog:
    def speak(self):
        print("Dog says: Woof!")

class Robot:
    def speak(self):
        print("Robot says: Beep Beep!")

def speak_object(obj):
    obj.speak()

dog = Dog()
robot = Robot()

speak_object(dog)
speak_object(robot)
```

```python
#15. Create Circle and Rectangle both having area().
#Concept
#Duck Typing
#What to do
#Function
#calculate(obj)
#How to solve
#1. Circle implements area().
#2. Rectangle implements area().
#3. Function calls area().
#4. Print returned value.

class Circle:
    def __init__(self, radius):
        self.radius = radius

    def area(self):
        return 3.14 * self.radius ** 2


class Rectangle:
    def __init__(self, length, width):
        self.length = length
        self.width = width

    def area(self):
        return self.length * self.width


def calculate(obj):
    print("Area:", obj.area())

circle = Circle(5)
rectangle = Rectangle(4, 6)

calculate(circle)
calculate(rectangle)
```

```python
#18. Create Logger classes using log(message).
#Concept
#Duck Typing
#What to do
#Two unrelated logger classes.
#How to solve
#1. FileLogger → log().
#2. ConsoleLogger → log().
#3. Function accepts logger object.
#4. Calls log(message).

class FileLogger:
    def log(self, message):
        print(f"File Log: {message}")

class ConsoleLogger:
    def log(self, message):
        print(f"Console Log: {message}")

def write_log(logger, message):
    logger.log(message)

file_logger = FileLogger()
console_logger = ConsoleLogger()

write_log(file_logger, "Application started")
write_log(console_logger, "Application started")
```

```python
#19. Create a hybrid inheritance structure and apply Duck Typing.
#Concept
#Hybrid + Duck Typing
#What to do
#Create hybrid inheritance.
#Create another unrelated class with same method.
#How to solve
#1. Create hybrid hierarchy.
#2. Final child has method show().
#3. Another unrelated class also has show().
#4. One function calls show() for both.

class Device:
    def power_on(self):
        print("Device is powered on.")

class Phone(Device):
    def call(self):
        print("Making a phone call.")

class Camera(Device):
    def click(self):
        print("Taking a photo.")

class SmartPhone(Phone, Camera):
    def show(self):
        print("This is a SmartPhone.")

class Laptop:
    def show(self):
        print("This is a Laptop.")

def display(obj):
    obj.show()

smartphone = SmartPhone()
laptop = Laptop()

smartphone.power_on()
smartphone.call()
smartphone.click()

display(smartphone)
display(laptop)
```

```python
#20. Create Employee hierarchy using Hybrid Inheritance and use Duck Typing.
#Concept
#Hybrid Inheritance + Duck Typing
#What to do
#Hierarchy
#Person
# |
#Employee
#Manager
#HR
#TeamLead
#Another class
#Freelancer
#Both TeamLead and Freelancer have
#work()
#How to solve
#1. Build hybrid inheritance.
#2. Final child defines work().
#3. Freelancer also defines work().
#4. Write a function:
#def do_work(obj):
#obj.work()
#5. Pass both objects to the function.
#6. Observe that the same function works for unrelated classes because of Duck Typing.


class Person:
    def introduce(self):
        print("I am a person.")

class Employee(Person):
    def employee_info(self):
        print("I am an employee.")

class Manager(Person):
    def manage(self):
        print("I manage the team.")

class TeamLead(Employee, Manager):
    def work(self):
        print("TeamLead is managing and coding.")

class Freelancer:
    def work(self):
        print("Freelancer is working on a client project.")

def do_work(obj):
    obj.work()

teamlead = TeamLead()
freelancer = Freelancer()

teamlead.introduce()
teamlead.employee_info()
teamlead.manage()

do_work(teamlead)
do_work(freelancer)
```
