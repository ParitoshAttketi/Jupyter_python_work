```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df= pd.read_csv("C:\\Users\\bhoom\\NARESH IT DATA SCIENCE\\Cars24_Used_car_data.csv")
df
```

```python
df.columns
```

```python
# Automatic vs Manual

df["Transmission"].value_counts().plot(kind="bar")
plt.title("Automatic vs Manual Cars")
plt.xlabel("Transmission")
plt.ylabel("Number of Cars")
plt.show()
```

```python
# Diesel vs Petrol

df["Fuel_Type"].value_counts().plot(kind="bar")
plt.title("Fuel Type")
plt.xlabel("Fuel Type")
plt.ylabel("Number of Cars")
plt.show()
```

```python
# mileage:

df.groupby("Fuel_Type")["Mileage_kmpl"].mean().plot(kind="bar")
plt.title("Average Mileage: Fuel Type")
plt.ylabel("Mileage (kmpl)")
plt.show()
```

```python
# number of cars by city:

df["Location"].value_counts().plot(kind="bar")
plt.title("Cars by City")
plt.xlabel("City")
plt.ylabel("Number of Cars")
plt.show()
```
