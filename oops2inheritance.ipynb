```python
# 
# duck typing
class Cd:
    def pay(self,amount):
        return f"paying amount using cd : {amount}"

class Dc:
    def pay(self,amount):
        return f"paying amount using Dc : {amount}"

class UPI:
    def pay(self,amount):
        return f"paying amount using UPI : {amount}"

#########
def make_payment(payment,amount):
    return payment.pay(amount)

print(make_payment(Cd(), 1000))
print(make_payment(Dc(), 2000))
print(make_payment(UPI(), 3000))


```

```python
class Circle:
    def draw(self):
        return "A circle"

    def area(self,radius):
        area = 3.14 * (radius**2)
        return f"Area of the circle : {area}"

class Rectangle:
    def draw(self):
        return "A Rectangle"

    def area(self,length,width):
        area = length * width
        return f"Area of Rectangle : {area}"

class Triangle:
    def draw(self):
        return "A Triangle"

    def area(self,base,hight):
        area = (1/2 * base * hight)
        return f"Area of Triangle : {area}"

def find_area(method, *args):
    return method.area(*args)

print(find_area(Circle(), 5))
print(find_area(Rectangle(), 10,11))  
print(find_area(Triangle(), 5,10))
```

```python
#Operator overloading

class cal:
    def __init__(self,grade):
        self.grade = grade
        
    # dunder method :__add__
    def __add__(self,other):
        result = self.grade + other.grade
        return result
    # dunder method :__sub__
    def __sub__(self,other):
        result = self.grade - other.grade
        return result
    # dunder method :__mul__
    def __mul__(self,other):
        result = self.grade * other.grade
        return result
    # dunder method :__truediv__
    def __truediv__(self,other):
        result = self.grade / other.grade
        return result

s1 = cal(80)
s2 = cal(50)

print(s1 + s2)
print(s1 - s2)
print(s1 * s2)
print(s1 / s2)
```
