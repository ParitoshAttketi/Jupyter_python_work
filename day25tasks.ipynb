```python
# 21. Create a custom exception InsufficientBalanceError for a banking program.

class InsufficientBalanceError(Exception):
    pass
balance = 5000

try:
    amount = int(input("enter amount"))
    if amount > balance:
        raise InsufficientBalanceError("Insufficient Balance please enter valid amount")
    balance -= amount
    print(f"total balance : {balance}")

except InsufficientBalanceError as e:
    print("error:",e)
```

```python
# 26. Create a simple calculator that handles ZeroDivisionError, ValueError and invalid operator.

try:
    num1 = float(input("Enter first number: "))
    op = input("Enter operator (+, -, *, /): ")
    num2 = float(input("Enter second number: "))

    if op == "+":
        print("Result:",num1 + num2)

    elif op == "-":
        print("Result:",num1 - num2)

    elif op == "*":
        print("Result:",num1 * num2)

    elif op == "/":
        print("Result:",num1 / num2)

    else:
        print("Invalid operator.")

except ZeroDivisionError:
    print("Error: Cannot divide by zero.")

except ValueError:
    print("Error: Please enter valid numbers.")
```

```python
# 30. Write a complete program that accepts user input for name, age and marks. Handle all
# possible exceptions and use finally to print “Program Ended”.

try:
    name = input("Enter your name: ")

    age = int(input("Enter your age: "))
    if age < 0:
        raise ValueError("Age cannot be negative.")

    marks = float(input("Enter your marks: "))
    if marks < 0 or marks > 100:
        raise ValueError("Marks must be between 0 and 100.")

    print("Student Details")
    print("Name :", name)
    print("Age :", age)
    print("Marks :", marks)

except ValueError as e:
    print("Error:", e)

except TypeError:
    print("Error: Invalid data type.")

except Exception as e:
    print("Unexpected Error:", e)

finally:
    print("Program Ended")
```
