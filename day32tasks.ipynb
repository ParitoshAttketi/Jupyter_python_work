```python
# 21.Extract the hour component from Duration and create Duration_Hour. Handle
# invalid or missing extracted values appropriately.

df['Duration_Hour'] = df['Duration'].str.split('h').str[0]
df['Duration_Hour'] = pd.to_numeric(df['Duration_Hour'], errors='coerce').fillna(0).astype(int)
df['Duration_Hour']
```

```python
# 22.Extract the minute component from Duration and create Duration_Minute.
# Handle invalid or missing extracted values appropriately.

df['Duration_Minute'] = df['Duration'].str.split('m').str[0]
df['Duration_Minute'] = pd.to_numeric(df['Duration_Minute'], errors='coerce').fillna(0).astype(int)
df['Duration_Minute'].value_counts()
```

```python
# 23.Drop the original Duration column after creating the numerical duration features.

df = df.drop(columns=['Duration'])
```

```python
# 24.Replace non-stop in Total_Stops with 0 stops.

df['Total_Stops'] = df['Total_Stops'].replace('non-stop', 0)
```

```python
# 25.Extract only the numerical value from Total_Stops and convert the column into
# an integer data type. Handle missing or invalid values appropriately.

df['Total_Stops'] = pd.to_numeric(df['Total_Stops'], errors='coerce').fillna(0).astype(int)
```
