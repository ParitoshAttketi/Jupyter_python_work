```python
# 1. Take a list of numbers. Using loop and operators, separate even and odd
# numbers, then calculate sum of even and product of odd numbers without
# using built-in sum() or prod().

numbers = list(map(int, input("Enter numbers separated by space: ").split()))

even = []
odd = []

even_sum = 0
odd_product = 1

for num in numbers:
    if num % 2 == 0:
        even.append(num)
        even_sum += num
    else:
        odd.append(num)
        odd_product *= num

if len(odd) == 0:
    odd_product = 0

print("Even Numbers:", even)
print("Odd Numbers:", odd)
print("Sum of Even Numbers:", even_sum)
print("Product of Odd Numbers:", odd_product)
```

```python
# 2. Create a program using nested loops and break/continue that prints all prime
# numbers between two user-given numbers, and also counts how many primes were
# found.

start = int(input("Enter the starting number: "))
end = int(input("Enter the ending number: "))

count = 0

print("Prime numbers are:")

for num in range(start, end + 1):

    if num < 2:
        continue

    for i in range(2, num):
        if num % i == 0:
            break
    else:
        print(num)
        count += 1

print("Total prime numbers found:", count)
```

```python
# 3. Take a string from user. Using string methods and loop, count vowels, consonants,
# digits and special characters separately.

text = input("Enter a string: ")

vowels = 0
consonants = 0
digits = 0
special = 0

for ch in text:
    if ch.lower() in "aeiou":
        vowels += 1
    elif ch.isalpha():
        consonants += 1
    elif ch.isdigit():
        digits += 1
    else:
        special += 1

print("Vowels:", vowels)
print("Consonants:", consonants)
print("Digits:", digits)
print("Special Characters:", special)
```

```python
# 4. Create a class BankAccount with private __balance.
# • Add deposit and withdraw with validation
# • Maintain a class attribute total_accounts
# • Add a class method to show total accounts
# • Overload + operator to transfer money from one account to another

class BankAccount:
    total_accounts = 0

    def __init__(self, name, balance):
        self.name = name
        self.__balance = balance
        BankAccount.total_accounts += 1

    def deposit(self, amount):
        if amount > 0:
            self.__balance += amount
            print("Deposited:", amount)
        else:
            print("Invalid deposit amount")

    def withdraw(self, amount):
        if amount <= 0:
            print("Invalid withdrawal amount")
        elif amount > self.__balance:
            print("Insufficient balance")
        else:
            self.__balance -= amount
            print("Withdrawn:", amount)

    def display(self):
        print(f"{self.name} Balance = ₹{self.__balance}")

    @classmethod
    def show_total_accounts(cls):
        print("Total Accounts:", cls.total_accounts)

    # Operator Overloading (+) for money transfer
    def __add__(self, other):
        amount = float(input("Enter amount to transfer: "))

        if amount <= 0:
            print("Invalid amount")
        elif amount > self.__balance:
            print("Insufficient balance")
        else:
            self.__balance -= amount
            other.__balance += amount
            print(f"₹{amount} transferred from {self.name} to {other.name}")

        return other

a1 = BankAccount("Rahul", 10000)
a2 = BankAccount("Priya", 5000)

a1.display()
a2.display()

a1.deposit(2000)
a2.withdraw(1000)

a1 + a2

a1.display()
a2.display()

BankAccount.show_total_accounts()
```

```python
# 5. Create abstract class Shape with abstract methods area() and perimeter().
# Implement Circle and Rectangle using super(). Add a class attribute to count total
# shapes created.

from abc import ABC, abstractmethod
import math

class Shape(ABC):
    total_shapes = 0

    def __init__(self):
        Shape.total_shapes += 1

    @abstractmethod
    def area(self):
        pass

    def perimeter(self):
        pass

    @classmethod
    def show_total_shapes(cls):
        print("Total Shapes Created:", cls.total_shapes)

class Circle(Shape):
    def __init__(self, radius):
        super().__init__()
        self.radius = radius

    def area(self):
        return math.pi * self.radius ** 2

    def perimeter(self):
        return 2 * math.pi * self.radius

class Rectangle(Shape):
    def __init__(self, length, width):
        super().__init__()
        self.length = length
        self.width = width

    def area(self):
        return self.length * self.width

    def perimeter(self):
        return 2 * (self.length + self.width)

c1 = Circle(5)
r1 = Rectangle(10, 4)

print("Circle Area:", round(c1.area(), 2))
print("Circle Perimeter:", round(c1.perimeter(), 2))

print("Rectangle Area:", r1.area())
print("Rectangle Perimeter:", r1.perimeter())

Shape.show_total_shapes()
```

```python
# 6. Create multi-level inheritance: Person → Employee → Manager.
# • Use private attributes
# • Add getter/setter with validation
# • Override a method at each level and use super()

class Person:
    def __init__(self, name, age):
        self.__name = name
        self.__age = age

    def get_name(self):
        return self.__name

    def get_age(self):
        return self.__age

    def set_name(self, name):
        if name.strip() != "":
            self.__name = name
        else:
            print("Invalid name")

    def set_age(self, age):
        if age > 0:
            self.__age = age
        else:
            print("Invalid age")

    def show_details(self):
        print("Name:", self.__name)
        print("Age:", self.__age)


class Employee(Person):
    def __init__(self, name, age, salary):
        super().__init__(name, age)
        self.__salary = salary

    def get_salary(self):
        return self.__salary

    def set_salary(self, salary):
        if salary > 0:
            self.__salary = salary
        else:
            print("Invalid salary")

    def show_details(self):
        super().show_details()
        print("Salary:", self.__salary)


class Manager(Employee):
    def __init__(self, name, age, salary, department):
        super().__init__(name, age, salary)
        self.__department = department

    def get_department(self):
        return self.__department

    def set_department(self, department):
        if department.strip() != "":
            self.__department = department
        else:
            print("Invalid department")

    def show_details(self):
        super().show_details()
        print("Department:", self.__department)

m = Manager("Rahul", 35, 75000, "IT")

print("Manager Details:")
m.show_details()

m.set_age(36)
m.set_salary(80000)
m.set_department("HR")

print("Updated Details:")
m.show_details()
```

```python
# 7. Create multiple inheritance with classes Camera and MusicPlayer. Create
# Smartphone that inherits from both. Resolve method name conflict using super() or
# MRO.

class Camera:
    def __init__(self):
        print("Camera Initialized")

    def start(self):
        print("Camera is Open")

    def info(self):
        print("This is Camera")


class MusicPlayer:
    def __init__(self):
        print("Music Player Initialized")

    def play_music(self):
        print("Playing Music")

    def info(self):
        print("This is Music Player")


class Smartphone(Camera, MusicPlayer):
    def __init__(self):
        super().__init__()

    def info(self):
        print("Smartphone Information")
        super().info()

phone = Smartphone()

phone.start()
phone.play_music()
phone.info()

print("Method Resolution Order (MRO):")
for cls in Smartphone.mro():
    print(cls.__name__)
```

```python
# 8. Create hybrid inheritance structure and demonstrate Method Resolution Order
# (MRO). Also apply Duck Typing on the final child class.

class Person:
    def work(self):
        print("Person is working")

class Employee(Person):
    def work(self):
        print("Employee is working")


class Student(Person):
    def study(self):
        print("Student is studying")

class Intern(Employee, Student):
    def work(self):
        print("Intern is working")

class Freelancer:
    def work(self):
        print("Freelancer is working")

def do_work(obj):
    obj.work()

i = Intern()
f = Freelancer()

print("Duck Typing:")
do_work(i)
do_work(f)

print("Method Resolution Order (MRO):")
for cls in Intern.mro():
    print(cls.__name__)
```

```python
# 9. Create a class ComplexNumber and overload +, -, *, and == operators completely.

class ComplexNumber:
    def __init__(self, real, imag):
        self.real = real
        self.imag = imag

    def __add__(self, other):
        return ComplexNumber(self.real + other.real,
                             self.imag + other.imag)

    def __sub__(self, other):
        return ComplexNumber(self.real - other.real,
                             self.imag - other.imag)

    def __mul__(self, other):
        real = self.real * other.real - self.imag * other.imag
        imag = self.real * other.imag + self.imag * other.real
        return ComplexNumber(real, imag)

    def __eq__(self, other):
        return self.real == other.real and self.imag == other.imag

    def __str__(self):
        if self.imag >= 0:
            return f"{self.real} + {self.imag}i"
        else:
            return f"{self.real} - {abs(self.imag)}i"

c1 = ComplexNumber(4, 5)
c2 = ComplexNumber(2, 3)

print("First Complex Number :", c1)
print("Second Complex Number:", c2)

print("Addition       :", c1 + c2)
print("Subtraction    :", c1 - c2)
print("Multiplication :", c1 * c2)

if c1 == c2:
    print("Both complex numbers are Equal")
else:
    print("Both complex numbers are Not Equal")
```

```python
# 10. Create a class Student with private marks list.
# • Add setter with validation (each mark 0-100)
# • Calculate average
# • Overload > operator to compare two students based on average

class Student:
    def __init__(self, name, marks):
        self.name = name
        self.__marks = []
        self.set_marks(marks)

    def set_marks(self, marks):
        valid = True

        for mark in marks:
            if mark < 0 or mark > 100:
                valid = False
                break

        if valid:
            self.__marks = marks
        else:
            print("Invalid marks! Each mark must be between 0 and 100.")

    def get_marks(self):
        return self.__marks

    def average(self):
        total = 0
        for mark in self.__marks:
            total += mark

        return total / len(self.__marks)

    def __gt__(self, other):
        return self.average() > other.average()

    def display(self):
        print("Name :", self.name)
        print("Marks:", self.__marks)
        print("Average:", round(self.average(), 2))

s1 = Student("Rahul", [80, 90, 85])
s2 = Student("Priya", [75, 95, 80])

print("Student 1 Details:")
s1.display()

print("Student 2 Details:")
s2.display()

print()

if s1 > s2:
    print(s1.name, "has a higher average.")
else:
    print(s2.name, "has a higher average.")
```

```python
# 11. Take a number from user. Using while loop, reverse the number and check if it is
# a palindrome (without converting to string).

num = int(input("Enter a number: "))

original = num
reverse = 0

while num > 0:
    digit = num % 10
    reverse = reverse * 10 + digit
    num = num // 10

print("Reversed Number:", reverse)

if original == reverse:
    print("Palindrome Number")
else:
    print("Not a Palindrome Number")
```

```python
# 12. Create hierarchical inheritance: Parent Account → Children SavingsAccount and CurrentAccount.


class Account(ABC):
    def __init__(self, acc_no, name, balance):
        self.acc_no = acc_no
        self.name = name
        self.balance = balance

    def display(self):
        print("Account No :", self.acc_no)
        print("Name       :", self.name)
        print("Balance    : ₹", self.balance)

    @abstractmethod
    def calculate_interest(self):
        pass

class SavingsAccount(Account):
    def __init__(self, acc_no, name, balance):
        super().__init__(acc_no, name, balance)

    def calculate_interest(self):
        interest = self.balance * 4 / 100    # 4% Interest
        print("Savings Interest: ₹", interest)

class CurrentAccount(Account):
    def __init__(self, acc_no, name, balance):
        super().__init__(acc_no, name, balance)

    def calculate_interest(self):
        interest = self.balance * 2 / 100    # 2% Interest
        print("Current Interest: ₹", interest)

s1 = SavingsAccount(101, "Rahul", 50000)
c1 = CurrentAccount(102, "Priya", 50000)

print("Savings Account")
s1.display()
s1.calculate_interest()

print("Current Account")
c1.display()
c1.calculate_interest()
```

```python
# 14. Using dictionary and loops, take student names and marks, store them, then find
# the student with highest and lowest marks without using max() or min().

students = {}

n = int(input("Enter the number of students: "))

for i in range(n):
    name = input("Enter student name: ")
    marks = int(input("Enter marks: "))
    students[name] = marks

first = True

for name, marks in students.items():
    if first:
        highest_name = lowest_name = name
        highest_marks = lowest_marks = marks
        first = False
    else:
        if marks > highest_marks:
            highest_marks = marks
            highest_name = name

        if marks < lowest_marks:
            lowest_marks = marks
            lowest_name = name

print("Student Records:")
for name, marks in students.items():
    print(name, ":", marks)

print("Highest Marks:")
print(highest_name, "-", highest_marks)

print("Lowest Marks:")
print(lowest_name, "-", lowest_marks)
```

```python
# 15. Create a class Time (hours, minutes, seconds) and overload + operator to add
# two time objects with proper carry-over.

class Time:
    def __init__(self, hours, minutes, seconds):
        self.hours = hours
        self.minutes = minutes
        self.seconds = seconds

    def __add__(self, other):
        sec = self.seconds + other.seconds
        carry_min = sec // 60
        sec = sec % 60

        minute = self.minutes + other.minutes + carry_min
        carry_hr = minute // 60
        minute = minute % 60

        hour = self.hours + other.hours + carry_hr

        return Time(hour, minute, sec)

    def __str__(self):
        return f"{self.hours:02}:{self.minutes:02}:{self.seconds:02}"

t1 = Time(2, 45, 50)
t2 = Time(3, 30, 25)

print("Time 1 :", t1)
print("Time 2 :", t2)

t3 = t1 + t2

print("Total Time :", t3)
```

```python
# 16. Create abstract class Payment with abstract method process(). Implement
# CreditCard and UPI. Use Duck Typing to process any payment object.

class Payment(ABC):

    @abstractmethod
    def process(self, amount):
        pass

class CreditCard(Payment):
    def process(self, amount):
        print(f"Credit Card Payment of ₹{amount} Successful.")

class UPI(Payment):
    def process(self, amount):
        print(f"UPI Payment of ₹{amount} Successful.")

class Wallet:
    def process(self, amount):
        print(f"Wallet Payment of ₹{amount} Successful.")

def make_payment(payment_obj, amount):
    payment_obj.process(amount)

cc = CreditCard()
upi = UPI()
wallet = Wallet()

print("Processing Payments:")
make_payment(cc, 5000)
make_payment(upi, 1200)
make_payment(wallet, 800)
```

```python
# 17. Create a class Employee with private salary.
# • Setter should not allow more than 20% increase at once
# • Keep history of salary changes using class attribute
# • Add static method to calculate tax


class Employee:
    salary_history = []

    def __init__(self, name, salary):
        self.name = name
        self.__salary = salary
        Employee.salary_history.append((self.name, salary))

    def get_salary(self):
        return self.__salary

    def set_salary(self, new_salary):
        max_salary = self.__salary * 1.20   # 20% increase allowed

        if new_salary <= max_salary:
            self.__salary = new_salary
            Employee.salary_history.append((self.name, new_salary))
            print("Salary Updated Successfully.")
        else:
            print("Error: Salary increase cannot exceed 20% at once.")

    def display(self):
        print("Employee Name :", self.name)
        print("Salary : ₹", self.__salary)

    @staticmethod
    def calculate_tax(salary):
        tax = salary * 0.10      # 10% tax
        return tax

e1 = Employee("Paritosh", 50000)

e1.display()

print("Updating Salary...")
e1.set_salary(58000)      # Allowed (16% increase)

e1.display()

print("Trying More Than 20% Increase...")
e1.set_salary(80000)      # Not Allowed

print("Salary History:")
for record in Employee.salary_history:
    print(record)

print("Tax on Current Salary: ₹",
      Employee.calculate_tax(e1.get_salary()))
```

```python
# 18. Write a program using nested loops that prints a diamond pattern of numbers
# and stops printing if user enters a negative number (use break).

while True:
    n = int(input("Enter the number of rows: "))

    if n < 0:
        print("Negative number entered. Program stopped.")
        break

    for i in range(1, n + 1):
        for j in range(n - i):
            print(" ", end=" ")

        for j in range(1, i + 1):
            print(j, end=" ")

        for j in range(i - 1, 0, -1):
            print(j, end=" ")

        print()

    for i in range(n - 1, 0, -1):
        for j in range(n - i):
            print(" ", end=" ")

        for j in range(1, i + 1):
            print(j, end=" ")

        for j in range(i - 1, 0, -1):
            print(j, end=" ")

        print()
```

```python
# 19. Create single, multi-level and hierarchical inheritance in one program and show
# the difference in method resolution.

class Animal:
    def sound(self):
        print("Animal makes a sound")

class Dog(Animal):
    def sound(self):
        print("Dog barks")

class Puppy(Dog):
    def sound(self):
        print("Puppy barks softly")

class Cat(Animal):
    def sound(self):
        print("Cat meows")


class Cow(Animal):
    def sound(self):
        print("Cow moos")

print("----- Single Inheritance -----")
d = Dog()
d.sound()

print("----- Multi-Level Inheritance -----")
p = Puppy()
p.sound()

print("----- Hierarchical Inheritance -----")
c = Cat()
cw = Cow()
c.sound()
cw.sound()

# Method Resolution Order (MRO)
print("Method Resolution Order (MRO)")
print("Dog   :", [cls.__name__ for cls in Dog.mro()])
print("Puppy :", [cls.__name__ for cls in Puppy.mro()])
print("Cat   :", [cls.__name__ for cls in Cat.mro()])
print("Cow   :", [cls.__name__ for cls in Cow.mro()])
```

```python
# 20. Create a complete system:
# • Abstract class Vehicle
# • Multiple inheritance with Electric and Fuel
# • Operator overloading for comparing vehicles based on speed
# • Duck Typing for start() method
# • Private attributes + getter/setter with validation

from abc import ABC, abstractmethod

class Vehicle(ABC):
    def __init__(self, name, speed):
        self.__name = name
        self.__speed = speed

    def get_name(self):
        return self.__name

    def get_speed(self):
        return self.__speed

    def set_name(self, name):
        if name.strip() != "":
            self.__name = name
        else:
            print("Invalid Vehicle Name")

    def set_speed(self, speed):
        if speed > 0:
            self.__speed = speed
        else:
            print("Invalid Speed")

    @abstractmethod
    def vehicle_type(self):
        pass

class Electric:
    def charge(self):
        print("Charging Electric Vehicle...")

class Fuel:
    def refuel(self):
        print("Refueling Fuel Vehicle...")

class HybridVehicle(Vehicle, Electric, Fuel):
    def __init__(self, name, speed):
        super().__init__(name, speed)

    def vehicle_type(self):
        print("Hybrid Vehicle")

    def start(self):
        print(self.get_name(), "started.")

    def __gt__(self, other):
        return self.get_speed() > other.get_speed()

class Bike:
    def start(self):
        print("Bike started.")

def start_vehicle(obj):
    obj.start()

v1 = HybridVehicle("Toyota Hybrid", 180)
v2 = HybridVehicle("Tesla Hybrid", 220)

v1.vehicle_type()
v1.charge()
v1.refuel()

print("Vehicle Details:")
print(v1.get_name(), "-", v1.get_speed(), "km/h")
print(v2.get_name(), "-", v2.get_speed(), "km/h")

if v1 > v2:
    print("\n", v1.get_name(), "is Faster")
else:
    print("\n", v2.get_name(), "is Faster")

print("Duck Typing:")
bike = Bike()
start_vehicle(v1)
start_vehicle(bike)

print("Updating Speed...")
v1.set_speed(200)
print("Updated Speed:", v1.get_speed(), "km/h")
```
