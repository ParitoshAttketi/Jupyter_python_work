```python
# 1. Load and Display the Dataset

df.head()
```

```python
#  2. Dataset Information

df.info()
```

```python
# 3. Find Missing Values

df.isna().sum()
```

```python
# 4. Find Duplicate Orders

df["Order_ID"].duplicated().sum()
```

```python
# 5. Remove Duplicate Orders

df = df.drop_duplicates(subset="Order_ID", keep="first")
```

```python
df.shape
```

```python
# 6. Calculate Revenue

df["Revenue"] = df["Quantity"] * df["Unit_Price"]
```

```python
df[["Quantity", "Unit_Price", "Revenue"]]
```

```python
# 7. Calculate Final Amount

df["Discount"] = df["Discount"].str.replace("%", "").astype(float)

df["Final_Amount"] = df["Revenue"] - (df["Revenue"] * df["Discount"] / 100)

df[["Revenue", "Discount", "Final_Amount"]]
```

```python
#  8. Find High-Value Orders

df[df["Final_Amount"] > 10000]
```

```python
#  9. City Analysis

df.groupby("City")["Final_Amount"].sum()
```

```python
# 10. Category Analysis

df.groupby("Category")["Revenue"].sum()
```

```python
# 11. Average Order Value by Payment Method

df.groupby("Payment_Method")["Final_Amount"].mean()
```

```python
#  12. Top 5 Products by Revenue

df.groupby("Product")["Revenue"].sum().sort_values(ascending=False).head(5)
```

```python
# 13 Most Ordered Category

df.groupby("Category")["Quantity"].sum().sort_values(ascending=False)
```

```python
# 14. Customer Spending

df.groupby("Customer_Name")["Final_Amount"].sum().sort_values(ascending=False).head(10)
```

```python
# 15. Rating Analysis

df.groupby("Category")["Customer_Rating"].mean()
```

```python
# 16. Returned Orders

df["Returned"].value_counts(normalize=True) * 100
```

```python
# 17. Date Analysis

df["Order_Date"] = pd.to_datetime(df["Order_Date"])

df["Month"] = df["Order_Date"].dt.month
```

```python
df[["Order_Date", "Month"]]
```

```python
# 18. Monthly Revenue

df.groupby("Month")["Final_Amount"].sum()
```

```python
df.groupby("Month")["Final_Amount"].sum().sort_index()
```

```python
# 19. Multiple Conditions

df[
    (df["Category"] == "Electronics") &
    (df["Final_Amount"] > 5000) &
    (df["Customer_Rating"] > 4)
]
```
