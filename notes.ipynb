```python
# try except else:

try:
    a = input("enter user id")
    b = input("enter password")
    
    if a == 123 and b == 122:
        print("username and password is correct")

except Exception as e:
    print(e)
else:
    print("login successfully")
```

```python
# try expect else finally

try:
    a = 20
    b = "10"
    c = a+b
except Exception as e:
    print(e)
else:
    print(c)
finally:
    print("all process done")
```
