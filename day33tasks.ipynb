```python
# 15.Identify all values in income_usd that cannot be directly interpreted as numbers.

for i in df["income_usd"].dropna().unique():
    try:
        float(i)
    except:
        print(i)
```

```python
# 16.Clean income_usd by removing currency symbols, commas, and invalid text,
# then convert it to a numeric column using pd.to_numeric().

df["income_usd"] = (df["income_usd"].astype(str).str.replace("$", "", regex=False).str.replace(",", "", regex=False))
df["income_usd"] = pd.to_numeric(df["income_usd"], errors="coerce")

df["income_usd"].dtype
```

```python
# 17.Clean the Age column so values such as "35 years" become numeric values.

df['age'].value_counts()
```

```python
df['age'] = df['age'].str.replace(' years', '')
df['age'] = pd.to_numeric(df['age'], errors='coerce')
df['age'].head()
```

```python
# 18.Identify unrealistic age values using NumPy/Pandas and replace them with NaN.

import numpy as np

df.loc[df["age"] > 100, "age"] = np.nan
df.loc[df["age"] < 0, "age"] = np.nan

df["age"].isna().sum()
```

```python
# 19.Clean the credit_score column by removing text such as "pts" and converting
# the column to numeric.

df['credit_score'] = df['credit_score'].str.replace('pts', '') 
df['credit_score'] = pd.to_numeric(df['credit_score'], errors='coerce') 
df['credit_score'].head()
```

```python
# 22.Detect suspicious values such as -999 and 99999 in tenure_years and satisfaction.

df[df['tenure_years'].isin([-999, 99999])]
df[df['satisfaction'].isin([-999, 99999])]
```

```python
# 23.Replace invalid/suspicious numerical values with NaN.

df['tenure_years'] = df['tenure_years'].replace([-999, 99999], np.nan)
df['satisfaction'] = df['satisfaction'].replace([-999, 99999], np.nan)
df[['tenure_years', 'satisfaction']].head()
```

```python
# 24.Fill missing numerical values using the median of each numerical column.

num_cols = df.select_dtypes(include='number').columns
df[num_cols] = df[num_cols].fillna(df[num_cols].median())
df[num_cols].isnull().sum()
```

```python
# 25.Fill missing categorical values using the mode of each categorical column.

cat_cols = df.select_dtypes(include='object').columns
for col in cat_cols:
    df[col] = df[col].fillna(df[col].mode()[0])
df[cat_cols].isnull().sum()
```

```python
# 26.Clean the signup_date column, which contains multiple date formats and
# invalid dates, and convert it to Pandas datetime.

df['signup_date'] = pd.to_datetime(df['signup_date'], errors='coerce')
df['signup_date'].head()
```

```python
# 27.Find rows where signup_date could not be converted into a valid date.

invalid_dates = df[df['signup_date'].isna()]
invalid_dates
```

```python
# 28.Clean the lead_source column by standardizing values such as Website,
# website, WEB, Referral, referral , Google Ads, and google_ads.

df['lead_source'] = df['lead_source'].str.strip().str.lower()
df['lead_source'] = df['lead_source'].replace({
    'web': 'website',
    'google_ads': 'google ads'})
df['lead_source'].value_counts()
```

```python
# 29.Create a final cleaned DataFrame containing:
# • cleaned column names
# • no duplicate IDs
# • standardized categorical values
# • numeric columns converted to numeric types
# • invalid values converted to NaN
# • missing values handled
# • detected outliers handled

clean_df = df.copy()
clean_df.columns = df.columns.str.strip().str.lower().str.replace(' ', '_')
```

```python
clean_df = clean_df.drop_duplicates(subset='customer_id')
num_cols = clean_df.select_dtypes(include='number').columns
for col in num_cols:
    clean_df[col] = pd.to_numeric(clean_df[col], errors='coerce')
```

```python
clean_df['tenure_years'] = clean_df['tenure_years'].replace([-999, 99999], np.nan)
clean_df['satisfaction'] = clean_df['satisfaction'].replace([-999, 99999], np.nan)
```

```python
clean_df['credit_score'] = clean_df['credit_score'].astype(str)
clean_df['credit_score'] = clean_df['credit_score'].str.replace('pts', '')
clean_df['credit_score'] = pd.to_numeric( clean_df['credit_score'], errors='coerce' )
```

```python
clean_df['lead_source'] = clean_df['lead_source'].str.strip().str.lower()
clean_df['lead_source'] = clean_df['lead_source'].replace({ 'web': 'website', 'google_ads': 'google ads' })
```

```python
clean_df['signup_date'] = pd.to_datetime( clean_df['signup_date'], errors='coerce' )
```

```python
num_cols = clean_df.select_dtypes(include='number').columns
for col in num_cols:
    clean_df[col] = clean_df[col].fillna(clean_df[col].median())
```

```python
cat_cols = clean_df.select_dtypes(include='object').columns
for col in cat_cols:
    clean_df[col] = clean_df[col].fillna(clean_df[col].mode()[0])
```

```python
for col in num_cols:
    Q1 = clean_df[col].quantile(0.25)
    Q3 = clean_df[col].quantile(0.75)
    IQR = Q3 - Q1 
    lower = Q1 - 1.5 * IQR 
    upper = Q3 + 1.5 * IQR 
    clean_df[col] = clean_df[col].clip(lower, upper)
```
