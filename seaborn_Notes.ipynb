```python
# import all 8 plots at a time:


fig, axes = plt.subplots(2,4, figsize=(20,10))

fig.suptitle("Categorical Plots - All 8 Types",
             fontsize = 18,
             fontweight="bold")

axes = axes.flatten()# multi D converted into single D

# strip plot
# hue : we can divide the data into diff cat like male and female 
sns.stripplot(data = tips,
             x = "day",
             y = "total_bill",
             hue = "sex",
             palette = "dark",
             ax=axes[0])

sns.swarmplot(data = tips,
             x = "day",
             y = "total_bill",
             hue = "sex",
             palette = "dark",
             ax=axes[1])

sns.boxplot(data = tips,
             x = "day",
             y = "total_bill",
             hue = "sex",
             palette = "dark",
             ax=axes[2])

sns.violinplot(data = tips,
             x = "day",
             y = "total_bill",
             hue = "sex",
             palette = "dark",
             ax=axes[3])

sns.pointplot(data = tips,
             x = "day",
             y = "total_bill",
             hue = "sex",
             palette = "dark",
             ax=axes[4])

sns.barplot(data = tips,
             x = "day",
             y = "total_bill",
             hue = "sex",
             palette = "dark",
             ax=axes[5])

sns.countplot(data=tips,
              x="day",
              hue="sex",
              palette="dark",
              ax=axes[6])

sns.boxenplot(data = tips,
             x = "day",
             y = "total_bill",
             hue = "sex",
             palette = "dark",
             ax=axes[7])
```

### relplot

```python
# relation plot
```

```python
# relation plot with scatter

sns.relplot(data = tips,
            x="total_bill",
            y="tip",
            hue="sex",
            size="size",
            style="smoker",
            palette=["cyan","red"],
            sizes=(20,200),
            alpha=0.7,
            height=6,
            aspect=1.3)
            
```

```python
# lineplot with relplot:

sns.relplot(data=flights,
            x="year",
            y="passengers",
            hue="month",
            kind="line",
            col="month",
            col_wrap=4,
            palette="tab20",
            height=3,
            aspect=1.2,
            linewidth=2
           )
plt.title("line plot month wise")
plt.show
```

# regplot()

```python
sns.regplot(data=tips,
            x="total_bill",
            y="tip",
            color="skyblue",
            scatter_kws={"alpha":1,
                     "color":"red"},
            line_kws={"color":"skyblue",
                      "linewidth":5,
                      "alpha":0.9})
```

```python
# polynomial regression (2nd order)

sns.regplot(data=tips,
            x="size",
            y="total_bill",
            order=2,
            color="red",
            scatter_kws={"alpha":0.5})
```
