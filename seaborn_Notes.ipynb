### part 1 : catplot()

```python
#  make a simple catplot:
sns.catplot(data = tips,
            x= "day",
            y= "total_bill",
            kind = "boxen")
```

```python
sns.catplot(data = tips,
            x= "day",
            y= "total_bill",
            kind = "point")
```

```python
sns.catplot(data = tips,
            x= "day",
            y= "total_bill",
            kind = "bar")
```

```python
sns.catplot(data = tips,
            x= "day",
            #y= "total_bill",
            kind = "count")
```

```python
sns.catplot(data = tips,
            x= "day",
            y= "total_bill",
            kind = "strip")
```

```python
sns.catplot(data = tips,
            x= "day",
            y= "total_bill",
            kind = "swarm")
```

```python
# import all 8 plots at a time:


fig, axes = plt.subplots(2,4, figsize=(20,10))

fig.suptitle("Categorical Plots - All 8 Types",
             fontsize = 18,
             fontweight="bold")

axes = axes.flatten()# multi D converted into single D

# strip plot
# hue : we can divide the data into diff cat like male and female 
sns.stripplot(data = tips,
             x = "day",
             y = "total_bill",
             hue = "sex",
             palette = "dark",
             ax=axes[0])

sns.swarmplot(data = tips,
             x = "day",
             y = "total_bill",
             hue = "sex",
             palette = "dark",
             ax=axes[1])

sns.boxplot(data = tips,
             x = "day",
             y = "total_bill",
             hue = "sex",
             palette = "dark",
             ax=axes[2])

sns.violinplot(data = tips,
             x = "day",
             y = "total_bill",
             hue = "sex",
             palette = "dark",
             ax=axes[3])

sns.pointplot(data = tips,
             x = "day",
             y = "total_bill",
             hue = "sex",
             palette = "dark",
             ax=axes[4])

sns.barplot(data = tips,
             x = "day",
             y = "total_bill",
             hue = "sex",
             palette = "dark",
             ax=axes[5])

sns.countplot(data=tips,
              x="day",
              hue="sex",
              palette="dark",
              ax=axes[6])

sns.boxenplot(data = tips,
             x = "day",
             y = "total_bill",
             hue = "sex",
             palette = "dark",
             ax=axes[7])
```
