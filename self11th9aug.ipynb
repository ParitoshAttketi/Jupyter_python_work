# 4. What is __init__()?

```python
class Student:
    def __init__(self, name):
        self.name = name

s1 = Student("Rahul")
print(s1.name)
```

# 5. What is self?

```python
class Student:
    def show(self):
        print("Hello")

s1 = Student()
s1.show()
```

# What is encapsulation?

```python
class Bank:
    def __init__(self):
        self.balance = 5000

    def show_balance(self):
        print(self.balance)
```

# What is inheritance?

```python
class Parent:
    def show(self):
        print("Parent")

class Child(Parent):
    pass

c = Child()
c.show()
```

# What is single inheritance?
# When one child class inherits from one parent class.

```python
class A:
    def show(self):
        print("A")

class B(A):
    pass
```

# What is multiple inheritance?

```python
class A:
    def show_a(self):
        print("A")

class B:
    def show_b(self):
        print("B")

class C(A, B):
    pass
```

# What is polymorphism?

```python
class Dog:
    def sound(self):
        print("Bark")

class Cat:
    def sound(self):
        print("Meow")
```

# What is method overriding?
## When a child class provides its own version of a parent method, it is called method overriding.

```python
class Parent:
    def show(self):
        print("Parent")

class Child(Parent):
    def show(self):
        print("Child")
```

# What is abstraction?
## abstraction means hiding unnecessary implementation details and showing only the important features.

```python
from abc import ABC, abstractmethod

class Animal(ABC):

    @abstractmethod
    def sound(self):
        pass
```

# What is a class variable?
## A class variable is shared by all objects of the class.

```python
class Student:
    school = "ABC School"

s1 = Student()
s2 = Student()

print(s1.school)
print(s2.school)
```

# What is an instance variable?
## An instance variable belongs to a particular object.

```python
class Student:
    def __init__(self, name):
        self.name = name

s1 = Student("Rahul")
s2 = Student("Amit")
```

# What is a method?
## A method is a function defined inside a class.

```python
class Student:
    def study(self):
        print("Student is studying")
```

# What is a static method?
## A static method does not require self or cls.

```python
class Calculator:

    @staticmethod
    def add(a, b):
        return a + b

print(Calculator.add(10, 20))
```

# What is a class method?
## A class method works with the class itself and uses cls.

```python
class Student:
    school = "ABC"

    @classmethod
    def show_school(cls):
        print(cls.school)

Student.show_school()
```

# Difference between class method, instance method, and static method?

```python
class Demo:

    def instance_method(self):
        print("Instance")

    @classmethod
    def class_method(cls):
        print("Class")

    @staticmethod
    def static_method():
        print("Static")
```
