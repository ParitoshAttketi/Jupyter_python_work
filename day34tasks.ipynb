```python
# 27.Detect outliers in Annual Income using the IQR method.

df["Annual_Income"]=pd.to_numeric(df["Annual_Income"],errors = "coerce")
```

```python
Q1 = df['Annual_Income'].quantile(0.25)
Q3 = df['Annual_Income'].quantile(0.75)
```

```python
IQR = Q3 - Q1

lower = Q1 - 1.5 * IQR
upper = Q3 + 1.5 * IQR

outliers = df[(df['Annual_Income'] < lower) | (df['Annual_Income'] > upper)]

outliers
```

```python
numeric_col = df.select_dtypes(include=["int","float"]).columns
for i in numeric_col:
    mean = df[i].mean()
    df[i]=df[i].fillna(mean)
```

```python
ct_col = df.select_dtypes(include=["object"]).columns
for i in ct_col:
    mode = df[i].mode()
    df[i]=df[i].fillna(mode[0])
```
