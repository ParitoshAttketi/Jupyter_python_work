```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = pd.read_csv("C:\\Users\\bhoom\\Downloads\\VideoGame_sales_cleaned.xls")
df
```

```python
df.columns
```

## Which platforms sold the most?

```python
df.groupby("Platform")["Global_Sales"].sum().sort_values(ascending=False).plot(kind="bar")

plt.xlabel("Platform")
plt.ylabel("Global Sales")
plt.title("Top 10 Platforms by Global Sales")
plt.show()
```

## Which genres dominate?

```python
df["Genre"].value_counts().plot(kind="bar")

plt.xlabel("Genre")
plt.ylabel("Number of Games")
plt.title("Games by Genre")
plt.show()
```

```python
df.groupby("Genre")["Global_Sales"].sum().sort_values(ascending=False).plot(kind="bar")

plt.xlabel("Genre")
plt.ylabel("Global Sales")
plt.title("Global Sales by Genre")
plt.show()
```

## Which publishers have highest sales?

```python
df.groupby("Publisher")["Global_Sales"].sum().sort_values(ascending=False).head(10).plot(kind="bar")

plt.xlabel("Publisher")
plt.ylabel("Global Sales")
plt.title("Top 10 Publishers by Global Sales")
plt.show()
```

## How have game sales changed over time?

```python
df.groupby("Year")["Global_Sales"].sum().plot(kind="line")

plt.xlabel("Year")
plt.ylabel("Global Sales")
plt.title("Game Sales Over Time")
plt.show()
```

## Which regions prefer which genres?

```python
df.groupby("Genre")[["NA_Sales", "EU_Sales", "JP_Sales"]].sum().plot(kind="bar")

plt.xlabel("Genre")
plt.ylabel("Sales")
plt.title("Regional Sales by Genre")
plt.show()
```
