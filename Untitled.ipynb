```python
#
import pickle

students = {"Name" : "paritosh",
            "age" : 23,
            "mob_no" : 9908014134
           }
with open("pickle.bin","wb") as file:
    pickle.dump(students,file)

print("file write sucess")
```

```python
# read bi file
with open("pickle.bin","rb") as file:
    r = pickle.load(file)
    print(r)

print("file read")
```
