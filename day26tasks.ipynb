```python
# 14. Count how many times a particular word appears in a file.

file = open("details.txt", "r")

count = 0

for line in file:
    words = line.split()
    for word in words:
        if word == "Student":
            count += 1

print(f"Student appears {count} times")

file.close()
```

```python
# 15. Read a file and display the longest word in it.

file = open("details.txt", "r")

data = file.read()
words = data.split()

longest = ""

for word in words:
    if len(word) > len(longest):
        longest = word

print("Longest word:", longest)

file.close()
```

```python
# 17. Store a list of numbers in a file, then read and calculate their sum.

file = open("list1.txt","w")

numbers = [1,2,3,4,5]

for num in numbers:
    file.write(str(num) + "\n")

file.close()

file = open("list1.txt","r")
sum = 0
for num in file:
    sum += int(num)

file.close()
print(f"sum = {sum}")
```

```python
# 25. Read a file containing student marks, calculate average, and write the result in another file.

file = open("marks.txt", "w")

file.write("99 \n98 \n97 \n95 \n90")

file.close()
```

```python
with open("marks.txt", "r")as file:
    print(file.read())
    file.close()
```

```python
file = open("marks.txt", "r")

total = 0
count = 0

for line in file:
    total += int(line)
    count += 1
file.close()

avg = total / count

result = open("marks.txt", "w")
result.write("Total marks :" + str(total) + "\n")
result.write("Average marks :" + str(avg))

result.close()

```

```python
# 26. Create a program that keeps a record of daily expenses in a file and can show total expenses.

file = open("text1.txt", "a")

n = int(input("Enter number of expenses: "))

for i in range(n):
    expense = float(input("Enter expense: "))
    file.write(str(expense) + "\n")

file.close()

file = open("text1.txt", "r")

total = 0

for line in file:
    total += float(line)

file.close()

print("Total Expenses =", total)
```

```python
# 29. Create a menu-driven program:
# • Write to file
# • Read file
# • Append to file
# • Exit

while True:
    print("\n     MENU     ")
    print("1. Write to File")
    print("2. Read File")
    print("3. Append to File")
    print("4. Exit")

    choice = int(input("Enter your choice: "))

    if choice == 1:
        file = open("exp_details.txt", "w")
        text = input("Enter text: ")
        file.write(text)
        file.close()

    elif choice == 2:
        file = open("exp_details.txt", "r")
        print(file.read())
        file.close()

    elif choice == 3:
        file = open("exp_details.txt", "a")
        text = input("Enter text to append: ")
        file.write("\n" + text)
        file.close()

    elif choice == 4:
        print("Exited.")
        break

    else:
        print("Invalid Choice!")
```

```python
# 30. Write a program that handles file exceptions (FileNotFoundError, PermissionError)
# properly while reading and writing files.

try:
    file = open("details.txt", "r")
    print(file.read())
    file.close()

except FileNotFoundError:
    print("Error: File not found.")

except PermissionError:
    print("Error: Permission denied.")

try:
    file = open("details.txt", "w")
    file.write("Hello Python")
    file.close()
    print("Data written successfully.")

except PermissionError:
    print("Error: Permission denied while writing.")

except Exception as e:
    print("Error:", e)
```
