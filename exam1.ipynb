```python
# 1. Custom Exceptions
# Create the following custom exceptions:
# • InsufficientBalanceError → Raised when withdrawal amount is greater than balance
# • InvalidAmountError → Raised when amount is zero or negative
# • AccountNotFoundError → Raised when account number does not exist
# • DailyLimitExceededError → Raised when daily withdrawal limit is crossed

class InsufficientBalanceError(Exception):
    pass

class InvalidAmountError(Exception):
    pass

class AccountNotFoundError(Exception):
    pass

class DailyLimitExceededError(Exception):
    pass

accounts = {100: 5000,200: 10000}
daily_limit = 25000

def withdraw(account_no, amount):
    if account_no not in accounts:
        raise AccountNotFoundError("Account number does not exist.")

    if amount <= 0:
        raise InvalidAmountError("Amount must be greater than zero.")

    if amount > daily_limit:
        raise DailyLimitExceededError("Daily withdrawal limit exceeded.")

    if amount > accounts[account_no]:
        raise InsufficientBalanceError("Insufficient balance.")

    accounts[account_no] -= amount
    print("Remaining Balance:", accounts[account_no])

try:
    account = int(input("Enter Account Number: "))
    amount = float(input("Enter Withdrawal Amount: "))
    withdraw(account, amount)

except AccountNotFoundError as e:
    print(e)

except InvalidAmountError as e:
    print(e)

except DailyLimitExceededError as e:
    print("error :",e)

except InsufficientBalanceError as e:
    print(e)

except ValueError:
    print("Please enter valid numeric values.")
```

```python
# 2. Abstract Class
# Create an abstract class Account with:
# • Abstract methods: deposit(amount), withdraw(amount), get_balance()
# • Concrete method: show_account_info()

from abc import ABC, abstractmethod

class Account(ABC):

    def __init__(self, acc_no, name, balance):
        self.__acc_no = acc_no
        self.name = name
        self.balance = balance

    def get_acc_no(self):
        return self.__acc_no

    @abstractmethod
    def deposit(self, amount):
        pass

    def withdraw(self, amount):
        pass

    def get_balance(self):
        pass

    def show_account_info(self):
        print("Account No :", self.get_acc_no())
        print("Name       :", self.name)
        print("Balance    :", self.balance)

class SavingsAccount(Account):

    def deposit(self, amount):
        self.balance += amount
        print("Amount Deposited Successfully.")

    def withdraw(self, amount):
        if amount <= self.balance:
            self.balance -= amount
            print("Amount Withdrawn Successfully.")
        else:
            print("Insufficient Balance.")

    def get_balance(self):
        return self.balance

a = SavingsAccount(1001, "Rahul", 10000)

a.show_account_info()
a.deposit(5000)
a.withdraw(3000)
print("Current Balance:", a.get_balance())
```

```python
# 3. Concrete Classes
# A. SavingsAccount (inherits from Account)
# • Private attributes: __account_no, __name, __balance, __daily_withdrawn
# • Daily withdrawal limit = ₹25,000
# • Minimum balance = ₹1,000
# • Interest rate = 4%
# B. CurrentAccount (inherits from Account)
# • Private attributes: __account_no, __name, __balance
# • Overdraft limit = ₹10,000
# • No interest

class Account(ABC):

    @abstractmethod
    def deposit(self, amount):
        pass

    def withdraw(self, amount):
        pass

    def get_balance(self):
        pass

    def show_account_info(self):
        print("Account Number :", self._Account__account_no)
        print("Name           :", self._Account__name)
        print("Balance        :", self.get_balance())

class SavingsAccount(Account):

    DAILY_LIMIT = 25000
    MIN_BALANCE = 1000
    INTEREST_RATE = 4

    def __init__(self, account_no, name, balance):
        self.__account_no = account_no
        self.__name = name
        self.__balance = balance
        self.__daily_withdrawn = 0

    def deposit(self, amount):
        self.__balance += amount
        print("Amount Deposited Successfully.")

    def withdraw(self, amount):
        if self.__daily_withdrawn + amount > self.DAILY_LIMIT:
            print("Daily Withdrawal Limit Exceeded.")
        elif self.__balance - amount < self.MIN_BALANCE:
            print("Minimum Balance Must Be ₹1000.")
        else:
            self.__balance -= amount
            self.__daily_withdrawn += amount
            print("Withdrawal Successful.")

    def get_balance(self):
        return self.__balance

    def show_account_info(self):
        print("Savings Account")
        print("Account No :", self.__account_no)
        print("Name       :", self.__name)
        print("Balance    :", self.__balance)
        print("Interest   :", self.INTEREST_RATE, "%")

class CurrentAccount(Account):

    OVERDRAFT_LIMIT = 10000

    def __init__(self, account_no, name, balance):
        self.__account_no = account_no
        self.__name = name
        self.__balance = balance

    def deposit(self, amount):
        self.__balance += amount
        print("Amount Deposited Successfully.")

    def withdraw(self, amount):
        if self.__balance + self.OVERDRAFT_LIMIT >= amount:
            self.__balance -= amount
            print("Withdrawal Successful.")
        else:
            print("Overdraft Limit Exceeded.")

    def get_balance(self):
        return self.__balance

    def show_account_info(self):
        print("Current Account")
        print("Account No :", self.__account_no)
        print("Name       :", self.__name)
        print("Balance    :", self.__balance)
        print("Interest   : No Interest")
        print("Overdraft  : ₹10000")

s = SavingsAccount(1001, "Rahul", 50000)
c = CurrentAccount(2001, "Amit", 8000)

s.deposit(5000)
s.withdraw(10000)
print("*"*20)
s.show_account_info()
print("*"*20)
c.deposit(2000)
c.withdraw(15000)
c.show_account_info()
```
