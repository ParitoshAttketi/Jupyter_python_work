```python
# Problem 7
# Create a custom iterator that prints only even numbers from 2 to 20.

class EvenNumbers:
    def __init__(self):
        self.num = 2

    def __iter__(self):
        return self

    def __next__(self):
        if self.num <= 20:
            value = self.num
            self.num += 2
            return value
        else:
            raise StopIteration

it = EvenNumbers()

for i in it:
    print(i)
```

```python
# Problem 23
# Create a custom iterator that prints all alphabet letters from A to Z.

class alph:
    def __init__(self):
        self.letter = "A"

    def __iter__(self):
        return self

    def __next__(self):
        if self.letter <= "Z":
            value = self.letter
            self.letter = chr(ord(self.letter) + 1)
            return value
        else:
            raise StopIteration

it = alph()

for i in it:
    print(i)
        
```

```python
# Problem 24
# Create a custom iterator that prints all vowels from the string below.
# text = "Artificial Intelligence"

# Problem 24
# Create a custom iterator that prints all vowels from the string below.
# text = "Artificial Intelligence"

text = "Artificial Intelligence"

class Vowels:
    def __init__(self, text):
        self.text = text
        self.index = 0

    def __iter__(self):
        return self

    def __next__(self):
        while self.index < len(self.text):
            ch = self.text[self.index]
            self.index += 1
            if ch.lower() in "aeiou":
                return ch
        raise StopIteration

it = Vowels(text)

for i in it:
    print(i)
```

```python
# 25 Create a custom iterator that prints the Fibonacci sequence up to 10 terms.

class Fibonacci:
    def __init__(self):
        self.a = 0
        self.b = 1
        self.count = 0

    def __iter__(self):
        return self

    def __next__(self):
        if self.count < 10:
            if self.count == 0:
                self.count += 1
                return self.a
            elif self.count == 1:
                self.count += 1
                return self.b
            else:
                self.a, self.b = self.b, self.a + self.b
                self.count += 1
                return self.b
        else:
            raise StopIteration

fib = Fibonacci()

for i in fib:
    print(i)
```

```python
# Problem 27
# Create a generator that yields all prime numbers between 1 and 100.

def prime_generator():
    for num in range(2, 101):
        prime = True

        for i in range(2, int(num ** 0.5) + 1):
            if num % i == 0:
                prime = False
                break

        if prime:
            yield num

for i in prime_generator():
    print(i, end=" ")
```

```python
# Problem 28
# Create a generator that yields all factors of a given number.
# Example Input:
# num = 24

def factors(num):
    for i in range(1, num + 1):
        if num % i == 0:
            yield i

num = 24

for i in factors(num):
    print(i, end=" ")
```
