```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

data = {
    "Name": ["Asha", "Rahul", "Priya", "Arjun", "Sneha",
             "Vijay", "Neha", "Kiran", "Ravi", "Anu",
             "Manoj", "Divya", "Rohit", "Pooja", "Suresh"],

    "Age": [20, 21, 19, 22, 20, 23, 21, 19, 22, 20, 
            24, 21, 23, 20, 22],

    "Gender": ["F", "M", "F", "M", "F", "M", "F", "M", 
               "M", "F", "M", "F", "M", "F", "M"],

    "Course": ["Python", "Python", "SQL", "Python", "SQL",
               "PowerBI", "Python", "SQL", "PowerBI", "Python",
               "SQL", "PowerBI", "Python", "SQL", "PowerBI"],

    "Marks": [85, 72, 91, 68, 88, 75, 95, 79, 82, 90,
              65, 87, 73, 92, 78],

    "Study_Hours": [5, 3, 6, 2, 5, 4, 7, 4, 5, 6,
                    2, 6, 3, 7, 4],

    "Attendance": [90, 75, 95, 70, 88, 80, 96, 78, 85, 92,
                   68, 90, 76, 94, 82],

    "City": ["Hyderabad", "Mumbai", "Delhi", "Hyderabad", "Chennai",
             "Mumbai", "Delhi", "Chennai", "Hyderabad", "Mumbai",
             "Delhi", "Chennai", "Hyderabad", "Delhi", "Mumbai"]
}

df = pd.DataFrame(data)
df
```

```python
# 1. Count Plot
# Create a count plot showing the number of students in each Course.

sns.countplot(data=data,
              x="Course",
              palette="Set2")
```

```python
# 2. Count Plot by Gender
#Create a count plot showing the number of students in each Gender.

sns.countplot(data=data,
              x="Gender",
              hue="Gender",
              palette="dark")
```

```python
# 3. Bar Plot
#Create a bar plot showing the average Marks for each Course.

sns.barplot(data=data,
            x="Course",
            y="Marks",
            hue="Course",
            palette="dark")
```

```python
# 4. Bar Plot by City
# Create a bar plot showing the average Marks for each City.

sns.barplot(data=data,
            x="City",
            y="Marks",
            hue="City",
            palette="dark",
            legend=False)
```

```python
# 5. Histogram
# Create a histogram showing the distribution of Marks.

sns.histplot(data=data,
             x="Marks",
             bins=5,
             kde=True,
             hue="Gender",
             palette="dark")
```

```python
# 6. Scatter Plot
# Create a scatter plot showing the relationship between Study_Hours and Marks.

sns.scatterplot(data=data,
                x="Study_Hours",
                y="Marks")
```

```python
# 7. Scatter Plot with Gender
#Create a scatter plot of Study_Hours vs Marks and use Gender as the hue.

sns.scatterplot(data=data,
                x="Study_Hours",
                y="Marks",
                hue="Gender")
```

```python
# 8. Scatter Plot with Course
# Create a scatter plot of Study_Hours vs Marks and use Course as the hue.

sns.scatterplot(data=data,
                x="Study_Hours",
                y="Marks",
                hue="Course")
```

```python
# 9. Line Plot
# Create a line plot showing Study_Hours vs Marks.


sns.lineplot(data=data,
             x="Study_Hours",
             y="Marks")
```

```python
# 10. Box Plot
# Create a box plot showing the distribution of Marks for each Course.

sns.boxplot(data=data,
            x="Course",
            y="Marks")
```

```python
# 11. Box Plot by Gender
# Create a box plot showing Marks for each Gender.

sns.boxplot(data=data,
            x="Gender",
            y="Marks")
```

```python
# 12. Violin Plot
# Create a violin plot showing Marks for each Course.

sns.violinplot(data=data,
               x="Course",
               y="Marks")
```

```python
# 13. Strip Plot
# Create a strip plot showing Marks for each Course.


sns.stripplot(data=data,
              x="Course",
              y="Marks")
```

```python
# 14. Swarm Plot
# Create a swarm plot showing Marks for each Course.

sns.swarmplot(data=data,
              x="Course",
              y="Marks")
```

```python
# 15. Bar Plot with Hue
# Create a bar plot showing the average Marks for each Course, with Gender as the hue.

sns.barplot(data=data,
            x="Course",
            y="Marks",
            hue="Gender")
```

```python
# 16. Study Hours by Course
# Create a box plot showing Study_Hours for each Course.

sns.boxplot(data=data,
            x="Course",
            y="Study_Hours")
```

```python
# 17. Attendance vs Marks
# Create a scatter plot showing the relationship between Attendance and Marks.

sns.scatterplot(data=data,
                x="Attendance",
                y="Marks")
```

```python
# 18. Attendance vs Marks with Gender
# Create a scatter plot of Attendance vs Marks, using Gender as hue.

sns.scatterplot(data=data,
                x="Attendance",
                y="Marks",
                hue="Gender")
```

```python
# 19. Correlation Heatmap
# Create a correlation heatmap for the numerical columns:
#Age
#Marks
#Study_Hours
#Attendance

import pandas as pd

data = pd.DataFrame(data)

sns.heatmap(data[["Age", "Marks", "Study_Hours", "Attendance"]].corr(),
            annot=True,
            cmap="coolwarm")
```

```python
# 20. Pair Plot
#Create a pair plot showing relationships between:
#Age
#Marks
#Study_Hours
#Attendance

sns.pairplot(data[["Age", "Marks", "Study_Hours", "Attendance", "Gender"]],
             hue="Gender")
```
