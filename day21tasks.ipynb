```python
# 4. Shape → Circle, Rectangle
# Question
# Define class Shape with method area(). Create subclasses Circle and Rectangle.
# Concept
# • Parent provides common method.
# • Children may later override it.

class Shape:
    def area(self):
        pass

class Circle(Shape):
    def __init__(self,radius):
        self.radius = radius
    def area(self):
        area = 3.14 * (self.radius**2)
        print(f"Area if the Circle : {area}")
class Rectangle(Shape):
    def __init__(self,length,width):
        self.length = length
        self.width = width

    def area(self):
        area = self.length * self.width
        print(f"Area of Rectangle : {area}")

c = Circle(5)
c.area()
x = Rectangle(11,12)
x.area()
```

```python
# 13. Person Constructor
# Concept
# Using super() to call the parent constructor.
# How to Solve
# 1. Create constructor in parent.
# 2. Create child constructors.
# 3. Use super().__init__().
# 4. Add child-specific attribute.
# 5. Display all data.


class Person:
    def __init__(self, name, age):
        self.name = name
        self.age = age


class Student(Person):
    def __init__(self, name, age, roll_no):
        super().__init__(name, age)   # Call parent constructor
        self.roll_no = roll_no        # Child-specific attribute

    def display(self):
        print(f"Name     : {self.name}")
        print(f"Age      : {self.age}")
        print(f"Roll No. : {self.roll_no}")


class Employee(Person):
    def __init__(self, name, age, salary):
        super().__init__(name, age)   
        self.salary = salary          

    def display(self):
        print(f"Name   : {self.name}")
        print(f"Age    : {self.age}")
        print(f"Salary : {self.salary}")

s = Student("Paritosh", 21, 101)
e = Employee("Rahul", 30, 50000)
print("="* 50)
print("Student Details")
s.display()
print("="* 50)
print("Employee Details")
e.display()
print("="* 50)
```

```python
# 16. Shape Area
# Concept
# Polymorphism using overriding.
# How to Solve
# 1. Parent constructor.
# 2. Circle calculates πr2.
# 3. Square calculates side2.
# 4. Override area().

class Shape:
    def area(self):
        print("Area depends on the shape")


class Circle(Shape):
    def __init__(self, radius):
        self.radius = radius

    def area(self):
        return 3.14 * self.radius ** 2


class Square(Shape):
    def __init__(self, side):
        self.side = side

    def area(self):
        return self.side ** 2

c = Circle(5)
s = Square(4)

print("Circle Area:", c.area())
print("Square Area:", s.area())
```

```python
# 23. School Management
# Concept
# Real-world hierarchical inheritance.
# How to Solve
# 1. Parent → Person
# 2. Common attributes (name, age)
# 3. Student → marks
# 4. Teacher → subject
# 5. Staff → department
# 6. Display information.

class Person:
    def __init__(self, name, age):
        self.name = name
        self.age = age

    def display(self):
        print("Name:", self.name)
        print("Age:", self.age)


class Student(Person):
    def __init__(self, name, age, marks):
        super().__init__(name, age)
        self.marks = marks

    def display(self):
        super().display()
        print("Marks:", self.marks)


class Teacher(Person):
    def __init__(self, name, age, subject):
        super().__init__(name, age)
        self.subject = subject

    def display(self):
        super().display()
        print("Subject:", self.subject)


class Staff(Person):
    def __init__(self, name, age, department):
        super().__init__(name, age)
        self.department = department

    def display(self):
        super().display()
        print("Department:", self.department)

student = Student("Paritosh", 22, 92)
teacher = Teacher("Anita", 35, "Mathematics")
staff = Staff("Ravi", 40, "Administration")
print("="*40)
print("Student Details")
student.display()
print("="*40)
print("Teacher Details")
teacher.display()
print("="*40)
print("Staff Details")
staff.display()
print("="*40)
```

```python
# 24. Bank System
# Concept
# Three children inherit one parent.
# How to Solve
# 1. Parent → Account
# 2. Store balance.
# 3. Savings → interest.
# 4. Current → overdraft.
# 5. FixedDeposit → maturity.
# 6. Show each separately.

class Account:
    def __init__(self, balance):
        self.balance = balance

    def show_balance(self):
        print("Balance:", self.balance)


class Savings(Account):
    def interest(self):
        print("Interest:", self.balance * 0.05)


class Current(Account):
    def overdraft(self):
        print("Overdraft Limit:", self.balance * 0.20)


class FixedDeposit(Account):
    def maturity(self):
        print("Maturity Amount:", self.balance * 1.08)

s = Savings(100000)
c = Current(100000)
f = FixedDeposit(100000)
print("="*40)
print("Savings Account")
s.show_balance()
s.interest()
print("="*40)
print("Current Account")
c.show_balance()
c.overdraft()
print("="*40)
print("Fixed Deposit Account")
f.show_balance()
f.maturity()
print("="*40)
```

```python
# 27. Employee Salary
# Concept
# Different salary calculations.
# How to Solve
# 1. Parent stores common data.
# 2. FullTime calculates monthly salary.
# 3. PartTime calculates hourly salary.
# 4. Override calculation method.

class Employee:
    def __init__(self, name):
        self.name = name

    def calculate_salary(self):
        print("Salary calculation depends on employee type.")


class FullTime(Employee):
    def __init__(self, name, monthly_salary):
        super().__init__(name)
        self.monthly_salary = monthly_salary

    def calculate_salary(self):
        print("Employee:", self.name)
        print("Monthly Salary:", self.monthly_salary)


class PartTime(Employee):
    def __init__(self, name, hours, rate_per_hour):
        super().__init__(name)
        self.hours = hours
        self.rate_per_hour = rate_per_hour

    def calculate_salary(self):
        salary = self.hours * self.rate_per_hour
        print("Employee:", self.name)
        print("Hourly Salary:", salary)

ft = FullTime("Paritosh", 50000)
pt = PartTime("Rahul", 80, 250)

ft.calculate_salary()
print("*"*30)
pt.calculate_salary()
```
