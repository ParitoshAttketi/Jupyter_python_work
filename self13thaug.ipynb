```python
#1. Find the second highest unique number

numbers = [10, 20, 40, 30, 40, 50, 20, 50]

unique_numbers = list(set(numbers))
unique_numbers.sort()

print("Second highest:", unique_numbers[-2])
```

```python
# 2. Count frequency of each word

sentence = "python is easy and python is powerful"

words = sentence.split()

frequency = {}

for word in words:
    if word in frequency:
        frequency[word] += 1
    else:
        frequency[word] = 1

print(frequency)
```

```python
#3. Find employees earning above average


employees = {
    "Rahul": 45000,
    "Priya": 60000,
    "Amit": 55000,
    "Sneha": 75000,
    "Ravi": 40000
}

average = sum(employees.values()) / len(employees)

print("Average salary:", average)

for name, salary in employees.items():
    if salary > average:
        print(name, salary)
```

```python
# 4. Find duplicate elements without using set()

numbers = [10, 20, 30, 20, 40, 10, 50, 30]

duplicates = []

for num in numbers:
    if numbers.count(num) > 1 and num not in duplicates:
        duplicates.append(num)

print(duplicates)
```

```python
# 5. Group students based on marks

students = {
    "Amit": 92,
    "Priya": 78,
    "Rahul": 65,
    "Sneha": 45,
    "Kiran": 88
}

result = {}

for name, marks in students.items():

    if marks >= 90:
        result[name] = "Excellent"

    elif marks >= 75:
        result[name] = "Good"

    elif marks >= 50:
        result[name] = "Average"

    else:
        result[name] = "Fail"

print(result)
```
