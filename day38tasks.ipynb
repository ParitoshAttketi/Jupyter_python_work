```python
# 12. Passenger Class vs Fare with Gender
#Create a categorical visualization comparing fare across passenger classes.
#Use:
#x="class"
#y="fare"
#hue="sex"
#Choose an appropriate kind.

sns.catplot(
    data=titanic,
    x="class",
    y="fare",
    hue="sex",
    kind="box"
)

plt.title("Fare Distribution by Passenger Class and Gender")
plt.xlabel("Passenger Class")
plt.ylabel("Fare")
```

```python
# 13. Survival vs Age Group
#Create an age_group column:
#• Child → Age < 18
#• Adult → 18–59
#• Senior → 60+
#Use catplot() to compare survival across these groups.

titanic["age_group"] = titanic["age"].apply(
    lambda age: "Child" if age < 18
    else "Adult" if age <= 59
    else "Senior"
)
sns.catplot(
    data=titanic,
    x="age_group",
    y="survived",
    kind="bar"
)

plt.title("Survival Rate by Age Group")
plt.xlabel("Age Group")
plt.ylabel("Survival Rate")
```

```python
# 14. Class, Gender and Survival
#Create a categorical visualization where:
#• X → class
#• Y → survival-related measure
#• Hue → sex
#• Columns → survived
#Use col= to create separate panels.

titanic = sns.load_dataset("titanic")

sns.catplot(
    data=titanic,
    x="class",
    y="survived",
    hue="sex",
    col="survived",
    kind="bar"
)

plt.show()
```

```python
# 15. Titanic Multi-Panel Analysis
#Create a catplot() with:
#• X → class
#• Y → fare
#• Hue → sex
#• Columns → survived
#Use an appropriate categorical kind.
#Add:
#• Figure size
#• Title
#• Color palette
#• Appropriate labels
#Explain what you observe.

g = sns.catplot(
    data=titanic,
    x="class",
    y="fare",
    hue="sex",
    col="survived",
    kind="box",
    height=5,
    aspect=1.2,
    palette="Set2")

g.set_axis_labels("Passenger Class", "Fare")
g.set_titles("Survived = {col_name}")
g.fig.suptitle("Fare Distribution by Class, Gender and Survival", y=1.05)
```
