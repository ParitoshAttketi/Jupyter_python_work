```python
# 1. Shopping Cart with Encapsulation

class Product:

    def __init__(self, name, price):
        self.name = name
        self.__price = price

    def get_price(self):
        return self.__price

    def set_price(self, price):
        if price > 0:
            self.__price = price
        else:
            print("Invalid price")


class ShoppingCart:

    def __init__(self):
        self.products = []

    def add_product(self, product):
        self.products.append(product)

    def remove_product(self, product):
        if product in self.products:
            self.products.remove(product)

    def total_price(self):
        total = 0

        for product in self.products:
            total += product.get_price()

        return total


p1 = Product("Laptop", 50000)
p2 = Product("Mouse", 1000)

cart = ShoppingCart()

cart.add_product(p1)
cart.add_product(p2)

print("Total:", cart.total_price())

p1.set_price(55000)

print("New Total:", cart.total_price())
```

```python
# 2. Bank Account System with Custom Exceptions

class InsufficientBalanceError(Exception):
    pass


class BankAccount:
    account_count = 0

    def __init__(self, name, balance):
        self.name = name
        self.balance = balance
        BankAccount.account_count += 1

    def deposit(self, amount):
        self.balance += amount
        print("Deposited:", amount)

    def withdraw(self, amount):
        if amount > self.balance:
            raise InsufficientBalanceError("Insufficient balance")
        
        self.balance -= amount
        print("Withdrawn:", amount)

    def display(self):
        print("Name:", self.name)
        print("Balance:", self.balance)

    @classmethod
    def total_accounts(cls):
        print("Total Accounts:", cls.account_count)


a1 = BankAccount("Rahul", 50000)
a2 = BankAccount("Amit", 30000)

a1.deposit(5000)
a1.withdraw(10000)

a1.display()

BankAccount.total_accounts()
```

```python
# 3. Employee Salary System Using Inheritance

class Employee:

    def __init__(self, name, salary):
        self.name = name
        self.salary = salary

    def calculate_salary(self):
        return self.salary


class Developer(Employee):

    def __init__(self, name, salary, coding_bonus):
        super().__init__(name, salary)
        self.coding_bonus = coding_bonus

    def calculate_salary(self):
        return self.salary + self.coding_bonus


class Manager(Employee):

    def __init__(self, name, salary, team_bonus):
        super().__init__(name, salary)
        self.team_bonus = team_bonus

    def calculate_salary(self):
        return self.salary + self.team_bonus


e1 = Developer("John", 60000, 10000)
e2 = Manager("David", 80000, 20000)

print(e1.name, e1.calculate_salary())
print(e2.name, e2.calculate_salary())
```

```python
# 5. Abstract Class for Payment System

from abc import ABC, abstractmethod


class Payment(ABC):

    @abstractmethod
    def pay(self, amount):
        pass


class CreditCardPayment(Payment):

    def pay(self, amount):
        print("Paid", amount, "using Credit Card")


class UPIPayment(Payment):

    def pay(self, amount):
        print("Paid", amount, "using UPI")


class CashPayment(Payment):

    def pay(self, amount):
        print("Paid", amount, "using Cash")


p1 = CreditCardPayment()
p2 = UPIPayment()
p3 = CashPayment()

p1.pay(5000)
p2.pay(2000)
p3.pay(1000)
```
