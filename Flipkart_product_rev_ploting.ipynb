```python
import seaborn as sns
import pandas as pd
import numpy as np
```

```python
df = pd.read_csv("C:\\Users\\bhoom\\NARESH IT DATA SCIENCE\\Flipkart_prod_rev_cleaned.csv")
df
```

```python
df.columns
```

```python
import matplotlib.pyplot as plt

plt.hist(df['Price'], bins=10)
plt.xlabel('Price')
plt.ylabel('Number of Products')
plt.title('Price Distribution')
plt.show()
```

```python
df['Rate'].value_counts().sort_index().plot(kind='bar')

plt.xlabel('Rating')
plt.ylabel('Number of Products')
plt.title('Rating Distribution')
plt.show()
```

```python
plt.scatter(df['Rate'], df['Price'])

plt.xlabel('Rating')
plt.ylabel('Price')
plt.title('Price vs Rating')
plt.show()
```

```python
avg_price = df.groupby('Rate')['Price'].mean()

avg_price.plot(kind='bar')

plt.xlabel('Rating')
plt.ylabel('Average Price')
plt.title('Average Price by Rating')
plt.show()
```

```python
top10_rating = df.sort_values('Rate', ascending=False).head(10)

plt.bar(top10_rating['Product_name'], top10_rating['Rate'])

plt.xlabel('Product')
plt.ylabel('Rating')
plt.title('Top 10 Products by Rating')
plt.xticks(rotation=90)
plt.show()
```
