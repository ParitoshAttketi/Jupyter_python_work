```python
#1. Create a Calculator class that can add 2 numbers or 3 numbers using the same method name.
#Concept
#Method Overloading (Python uses default arguments or *args to achieve overloading.)
#What to do
#• Create a class Calculator.
#• Create one method add().
#• The method should work for either 2 numbers or 3 numbers.
#How to solve
#1. Create class Calculator.
#2. Define method add().
#3. Use default arguments or *args.
#4. Check how many numbers are provided.
#5. Return the sum.

class Calculator:
    def __add__(self,a,b,c=0,d=0):
        result = a+b+c+d
        return result

x = Calculator()

x.__add__(1,2,3,4) 
```

```python
# 2. Create an Area class that calculates the area of a square and rectangle using the same
#method.
#Concept
#Method Overloading using default arguments.
#What to do
#• One parameter → Square.
#• Two parameters → Rectangle.
#How to solve
#1. Create class Area.
#2. Define method calculate().
#3. If only one value is provided, calculate the square's area.
#4. If two values are provided, calculate the rectangle's area.
#5. Return the result.

class Area:
    def calculate(self,a,b=None):
        if b is None:
            return f"Square's area : {a*a}"
        else:
            return f"Area of rectangle : {a*b}"

x = Area()
print(x.calculate(10))
print(x.calculate(10,11))
```

```python
# 3. Create a Student class with a details() method that accepts different numbers of arguments.
#Concept
#Method Overloading
#What to do
#The method should work with:
#• Name
#• Name + Age
#• Name + Age + City
#How to solve
#1. Create class Student.
#2. Create method details().
#3. Use default arguments.
#4. Print the information based on the provided values.

class Student:
    def details(self,name,age=None,city=None):
        self.name = name
        if age is None and city is None:
            return f"Name of the student : {self.name}"
        elif age is not None and city is None:
            print(f"Name of the student : {self.name}")
            print(f"Age of the student : {age}")
        elif age is None and city is not None:
            print(f"Name of the student : {self.name}")
            print(f"City of the student : {city}")
        else:
            print(f"Name of the student : {self.name}")
            print(f"Age of the student : {age}")
            print(f"City of the student : {city}")
x = Student()
x.details("paritosh",None,"hyderabad")
```

```python
# 4. Create a MathOperations class that multiplies any number of values.
#Concept
#Method Overloading using *args
#What to do
#Multiply:
#• 2 numbers
#• 3 numbers
#• 4 numbers
#• More numbers
#How to solve
#1. Create class MathOperations.
#2. Define method multiply(*args).
#3. Initialize the answer as 1.
#4. Loop through all numbers.
#5. Multiply each number.
#6. Return the final product.

class MathOperations:
    def multiply(self,*args):
        answer = 1

        for num in args:
            answer *= num 
        
        return answer

x = MathOperations()
print(x.multiply(2,3))
print(x.multiply(2,3,5))
print(x.multiply(2,3,5,9))
```

```python
# 5. Create a Display class that prints different messages based on the argument type.
# Concept
# Method Overloading + Type Checking
# What to do
#Accept:
#• Integer
#• String
#• List
#How to solve
#1. Create class Display.
#2. Create method show().
#3. Use isinstance().
#4. Print different messages depending on the object's type.

class Display:
    def show(self,val):
        if isinstance(val,int):
            return f"Integer value : {val}"

        elif isinstance(val,str):
            return f"String value : {val}"
            
        elif isinstance(val,list):
            return f"List value : {val}"

        else:
            return "not in any form"

x = Display()
print(x.show(1))
print(x.show("paritosh"))
print(x.show([1,2,3]))
print(x.show(1.2))
    
```

```python
# 6. Create a Volume class that calculates the volume of a cube and cuboid.
# Concept
# Method Overloading
#What to do
#• One parameter → Cube
#• Three parameters → Cuboid
#How to solve
#1. Create class Volume.
#2. Create method calculate().
#3. If one value is given, calculate the cube's volume.
#4. Otherwise, calculate the cuboid's volume.

class Volume:
    def calculate(self,a,b=None,c=None):
        if  b is None and c is None:
            return f"cube's volume : {a**3}"
        else:
            return f"cuboid's volume : {a*b*c}"

x = Volume()
x.calculate(2)
```

```python
# 7. Create a PrintData class that prints one or many values.
# Concept
#Method Overloading using *args
#What to do
#Print:
#• One value
#• Multiple values
#How to solve
#1. Create class PrintData.
#2. Create method print_data(*args).
#3. Loop through the arguments.
#4. Print each value.

class PrintData:
    def print_data(self,*args):
        for i in args:
            print(i)

x = PrintData()
x.print_data("paritosh","sai","santosh")
```

```python
# 8. Create an Animal class with sound(). Override it in Dog and Cat.
#Concept
#Method Overriding
#What to do
#• Animal → Generic sound
#• Dog → Bark
#• Cat → Meow
#How to solve
#1. Create parent class Animal.
#2. Define method sound().
#3. Inherit Dog and Cat.
#4. Override sound().
#5. Create objects and call the method.

class Animal:
    def sound(self):
        print("Animal sound")

class Dog(Animal):
    def sound(self):
        print("Dog barks")

class Cat(Animal):
    def sound(self):
        print("meow")

x = Dog()
x.sound()
z= Cat()
z.sound()
```

```python
# 9. Create Vehicle, Car, and Bike classes.
# Concept
#Method Overriding
#What to do
#Override the start() method.
#How to solve
#1. Create Vehicle.
#2. Add method start().
#3. Create Car.
#4. Create Bike.
#5. Override start() in both child classes.

class Vehicle:
    def start(self):
        pass

class Car(Vehicle):
    def start(self):
        print("car is a Vehicle")

class Bike(Vehicle):
    def start(self):
        print("Bike is a Vehicle")

c= Car()
c.start()

v=Bike()
v.start()
```

```python
# 10. Create an Employee class and override calculate_salary() in Manager.
#Concept
#Method Overriding
#What to do
#• Employee → Normal salary
#• Manager → Salary + Bonus
#How to solve
#1. Create parent class.
#2. Define calculate_salary().
#3. Inherit Manager.
#4. Override the method.
#5. Print the manager's salary.

class Employee:
    def __init__(self,salary):
        self.salary = salary
        
    def calculate_salary(self):
        return f"Employee salary : {self.salary}"

class Manager(Employee):
    def __init__(self,salary,bonus):
        super().__init__(salary)
        self.bonus = bonus
        
    def Bonus(self):
        total_salary = self.salary + self.bonus
        return f"Manager Total salary : {total_salary}"

d = Manager(10000,500)
d.Bonus()
        
```

```python
#11. Create Shape, Circle, and Rectangle classes.
#Concept
#Method Overriding
#What to do
#Override the area() method.
#How to solve
#1. Create Shape.
#2. Add area().
#3. Inherit Circle.
#4. Inherit Rectangle.
#5. Override area() in each child class.

class Shape:
    def area(self):
        pass

class Circle(Shape):
    def area(self,radius):
        self.radius = radius
        area = 3.14 * (radius**2)
        return f"Area of Circle : {area}"
        
class Rectangle(Shape):
    def area(self,length,width):
        self.length = length
        self.width = width
        area = self.length * self.width
        return f"Area of Rectangle : {area}"

x = Circle()
print(x.area(4))

c=Rectangle()
print(c.area(4,5))
```

```python
# 13. Create Parent and Child classes using super().
# Concept
# Method Overriding + super()
# What to do
# Override display().
# How to solve
# 1. Create Parent.
# 2. Add display().
# 3. Create Child.
# 4. Override display().
# 5. Call super().display().
# 6. Print additional information.

class Parent:
    def display(self):
        print("hello from parent class")

class Child(Parent):
    def display(self):
        super().display()
        print("hello from child class")

x = Child()
x.display()
```

```python
# 15. Create a Point class and overload the + operator.
# Concept
# Operator Overloading (__add__)
# What to do
# Add two points together.
# How to solve
# 1. Create class Point.
# 2. Store x and y coordinates.
# 3. Overload __add__().
# 4. Return a new point with added coordinates.

class Point:
    def __init__(self,x,y):
        self.x = x
        self.y = y

    def __add__(self):
        return self.x + self.y

x= Point(20,30)
x.__add__()
```

```python
# 16. Create a Book class and overload the + operator.
# Concept
# Operator Overloading
# What to do
# Add the number of pages of two books.
# How to solve
# 1. Create class Book.
# 2. Store pages.
# 3. Overload __add__().
# 4. Return the total pages.

class Book:
    def __init__(self,pages):
        self.pages = pages

    def __add__(self,new):
        total_pages = self.pages + new
        return f"Total pages : {total_pages}"

c = Book(30)
print(c.__add__(10))
```

```python
# 17. Create a Student class and overload the > operator.
# Concept
# Comparison Operator Overloading (__gt__)
# What to do
# Compare marks of two students.
# How to solve
# 1. Create class Student.
# 2. Store marks.
# 3. Overload __gt__().
# 4. Compare the marks.
# 5. Return True or False.

class Student:
    def __init__(self,marks):
        self.marks = marks

    def __gt__(self,new):
        if self.marks == new:
            return True
        else:
            return False

x= Student(99)
x.__gt__(990)
```

```python
#18. Create a Rectangle class and overload the == operator.
# Concept
# Equality Operator Overloading (__eq__)
# What to do
# Compare rectangles using their area.
# How to solve
# 1. Create class Rectangle.
# 2. Store length and width.
# 3. Calculate area.
# 4. Overload __eq__().
# 5. Compare the areas.

class Rectangle:
    def __init__(self,length,width):
        self.length = length
        self.width = width

    def area(self):
        area = self.length * self.width
        return f"Area of Rectangle : {area}"

    def __eq__(self, other):
        return self.area() == other.area()

r1 = Rectangle(4, 5)
r2 = Rectangle(2, 10)

print("Area of Rectangle 1:", r1.area())
print("Area of Rectangle 2:", r2.area())
print("Are both rectangles ", r1 == r2)
```

```python
#19. Create a Money class and overload the + and - operators.
#Concept
#Multiple Operator Overloading
#What to do
#Perform addition and subtraction of money amounts.
#How to solve
#1. Create class Money.
#2. Store the amount.
#3. Overload __add__().
#4. Overload __sub__().
#5. Return new Money objects after each operation.

class Money:
    def __init__(self,amount):
        self.amount = amount

    def __add__(self,new):
        total = self.amount + new
        return total

    def __sub__(self,new):
        total = self.amount - new
        return total

c = Money(200)
c.__add__(300)
c.__sub__(100)
        
```

```python
# 20. Create a Distance class and overload <, >, and ==.
#Concept
#Comparison Operator Overloading
#What to do
#Compare two distance objects.
#How to solve
#1. Create class Distance.
#2. Store the distance value.
#3. Overload:
#o __lt__() for <
#o __gt__() for >
#o __eq__() for ==
#4. Compare the stored distance values.
#5. Return True or False.

class Distance:
    def __init__(self, value):
        self.value = value

    def __lt__(self, other):
        return self.value < other.value

    def __gt__(self, other):
        return self.value > other.value

    def __eq__(self, other):
        return self.value == other.value

d1 = Distance(50)
d2 = Distance(75)

print("d1 < d2 :", d1 < d2)
print("d1 > d2 :", d1 > d2)
print("d1 == d2 :", d1 == d2)
```
