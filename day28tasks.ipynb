```python
# 14. Create class Rectangle with private length and width. Add method to calculate
# area and perimeter.

class Rectangle:
    def __init__(self,length,width):
        self.__length = length
        self.__width = width

    def area(self):
        area = self.__length * self.__width
        print(f"Area of Rectangle: {area}")

    def perimeter(self):
        per = 2 * (self.__length * self.__width)
        print(f"perimeter of Rectangle: {per}")

r = Rectangle(9,8)
r.area()
r.perimeter()
```

```python
# 18. Create abstract class Shape with abstract method area(). Implement it in Circle
# and Square using super().

from abc import ABC, abstractmethod

class Shape(ABC):
    def __init__(self, value):
        self.value = value

    @abstractmethod
    def area(self):
        pass

class Circle(Shape):
    def __init__(self, radius):
        super().__init__(radius)

    def area(self):
        return 3.14 * self.value ** 2

class Square(Shape):
    def __init__(self, side):
        super().__init__(side)

    def area(self):
        return self.value ** 2

c = Circle(7)
s = Square(5)

print("Area of Circle =", c.area())
print("Area of Square =", s.area())
```

```python
# 19. Create multiple inheritance with two classes Father and Mother. Create child
# class and resolve method conflict.

class Father:
    def __init__(self, father_name):
        self.father_name = father_name

    def introduce(self):
        print("I am Father")


class Mother:
    def __init__(self, mother_name):
        self.mother_name = mother_name

    def introduce(self):
        print("I am Mother")


class Child(Father, Mother):
    def __init__(self, father_name, mother_name, child_name):
        Father.__init__(self, father_name)
        Mother.__init__(self, mother_name)
        self.child_name = child_name

    def introduce(self):
        Father.introduce(self)

    def info(self):
        print("Father Name :", self.father_name)
        print("Mother Name :", self.mother_name)
        print("Child Name  :", self.child_name)


c = Child("Jack", "Akane", "Kai")
c.introduce()
c.info()
```

```python
# 26. Create a class Student with private marks list.
# • Setter should validate each mark (0-100)
# • Overload > operator to compare two students based on average
# • Raise custom exception if marks list is empty

class EmptyMarksError(Exception):
    pass


class Student:
    def __init__(self, name):
        self.name = name
        self.__marks = []

    def set_marks(self, marks):
        if len(marks) == 0:
            raise EmptyMarksError("Marks list cannot be empty.")

        for mark in marks:
            if mark < 0 or mark > 100:
                raise ValueError("Marks must be between 0 and 100.")

        self.__marks = marks

    def get_marks(self):
        return self.__marks

    def average(self):
        return sum(self.__marks) / len(self.__marks)

    def __gt__(self, other):
        return self.average() > other.average()

try:
    s1 = Student("paritosh")
    s2 = Student("santosh")

    s1.set_marks([80, 90, 85])
    s2.set_marks([75, 88, 82])

    print("paritosh Average:", s1.average())
    print("santosh Average:", s2.average())

    if s1 > s2:
        print("paritosh has a higher average.")
    else:
        print("santosh has a higher average.")

except EmptyMarksError as e:
    print("Custom Exception:", e)

except ValueError as e:
    print("ValueError:", e)
```

```python
# 28. Using dictionary + exception handling, create a student management system that can:
# • Add student
# • Update marks
# • Delete student
# • Display all Raise custom exception if student not found.

class StudentNotFoundError(Exception):
    pass

students = {}

while True:
    print("1. Add Student")
    print("2. Update Marks")
    print("3. Delete Student")
    print("4. Display All Students")
    print("5. Exit")

    try:
        choice = int(input("Enter your choice: "))

        if choice == 1:
            name = input("Enter student name: ")
            marks = int(input("Enter marks: "))
            students[name] = marks

        elif choice == 2:
            name = input("Enter student name to update: ")

            if name not in students:
                raise StudentNotFoundError("Student not found!")

            marks = int(input("Enter new marks: "))
            students[name] = marks

        elif choice == 3:
            name = input("Enter student name to delete: ")

            if name not in students:
                raise StudentNotFoundError("Student not found!")

            del students[name]

        elif choice == 4:
            if not students:
                print("No student records available.")
            else:
                print("Student Records")
                print("-"*30)
                for name, marks in students.items():
                    print(f"{name} : {marks}")

        elif choice == 5:
            print("Exiting program...")
            break
        else:
            print("Invalid choice!")

    except ValueError:
        print("Please enter a valid number.")

    except StudentNotFoundError as e:
        print(e)

    except Exception as e:
        print("Error:", e)
```

```python
# 30. Design a complete system combining:
# • Abstract class Vehicle
# • Multiple inheritance (Electric + Fuel)
# • Operator overloading (compare by speed)
# • Private attributes + getter/setter
# • Custom exceptions
# • File handling to save vehicle details
# • Duck Typing for start() method


from abc import ABC, abstractmethod

class SpeedError(Exception):
    pass

# Abstract Class
class Vehicle(ABC):
    def __init__(self, speed):
        self.set_speed(speed)

    def get_speed(self):
        return self.__speed

    def set_speed(self, speed):
        if speed < 0:
            raise SpeedError("Invalid Speed")
        self.__speed = speed

    @abstractmethod
    def drive(self):
        pass

    def __gt__(self, other):
        return self.get_speed() > other.get_speed()

class Electric:
    pass

class Fuel:
    pass


class Car(Vehicle, Electric, Fuel):
    def drive(self):
        print("Car is driving")

    def save(self):
        with open("vehicle.txt", "a") as f:
            f.write(f"Speed = {self.get_speed()}\n")

class Bike:
    def start(self):
        print("Bike Started")

class Bus:
    def start(self):
        print("Bus Started")

def begin(x):
    x.start()

try:
    c1 = Car(120)
    c2 = Car(90)

    c1.drive()

    if c1 > c2:
        print("Car 1 is Faster")
    else:
        print("Car 2 is Faster")

    c1.save()

    begin(Bike())
    begin(Bus())

except SpeedError as e:
    print(e)
```
