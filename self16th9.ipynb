```python
# 1. Student Result Management

class Student:
    def __init__(self, name, roll_no, marks):
        self.name = name
        self.roll_no = roll_no
        self.marks = marks

    def total(self):
        return sum(self.marks)

    def average(self):
        return self.total() / len(self.marks)

    def result(self):
        if self.average() >= 40:
            return "Pass"
        else:
            return "Fail"

    def display(self):
        print("Name:", self.name)
        print("Roll No:", self.roll_no)
        print("Total:", self.total())
        print("Average:", self.average())
        print("Result:", self.result())


s1 = Student("Rahul", 101, [75, 60, 80])
s1.display()
```

```python
# 2. Bank Account

class BankAccount:
    def __init__(self, name, account_no, balance):
        self.name = name
        self.account_no = account_no
        self.balance = balance

    def deposit(self, amount):
        self.balance += amount
        print("Amount deposited")

    def withdraw(self, amount):
        if amount <= self.balance:
            self.balance -= amount
            print("Amount withdrawn")
        else:
            print("Insufficient balance")

    def display_balance(self):
        print("Balance:", self.balance)


account = BankAccount("Ravi", 12345, 10000)

account.deposit(5000)
account.withdraw(3000)
account.display_balance()

```

```python
# 3. Employee Salary System


class Employee:
    def __init__(self, name, salary):
        self.name = name
        self.salary = salary

    def display_salary(self):
        print("Salary:", self.salary)


class Manager(Employee):
    def __init__(self, name, salary, bonus):
        super().__init__(name, salary)
        self.bonus = bonus

    def display_salary(self):
        print("Name:", self.name)
        print("Basic Salary:", self.salary)
        print("Bonus:", self.bonus)
        print("Final Salary:", self.salary + self.bonus)


m1 = Manager("Arjun", 50000, 10000)
m1.display_salary()
```

```python
# 4. Shopping Cart

class Product:
    def __init__(self, name, price):
        self.name = name
        self.price = price


class ShoppingCart:
    def __init__(self):
        self.products = []

    def add_product(self, product):
        self.products.append(product)

    def total_price(self):
        total = 0

        for product in self.products:
            total += product.price

        return total

    def display(self):
        for product in self.products:
            print(product.name, product.price)

        print("Total:", self.total_price())


p1 = Product("Laptop", 50000)
p2 = Product("Mouse", 1000)
p3 = Product("Keyboard", 2000)

cart = ShoppingCart()

cart.add_product(p1)
cart.add_product(p2)
cart.add_product(p3)

cart.display()
```

```python
# 5. Vehicle Rental System


class Vehicle:
    def __init__(self, name, price_per_day):
        self.name = name
        self.price_per_day = price_per_day

    def rental_cost(self, days):
        return self.price_per_day * days


class Car(Vehicle):
    def rental_cost(self, days):
        return self.price_per_day * days


class Bike(Vehicle):
    def rental_cost(self, days):
        return self.price_per_day * days


car = Car("Toyota", 2000)
bike = Bike("Honda", 800)

print("Car rental:", car.rental_cost(5))
print("Bike rental:", bike.rental_cost(5))
```
